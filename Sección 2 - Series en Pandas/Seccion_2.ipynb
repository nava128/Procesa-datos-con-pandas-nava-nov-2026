### ¿Qué son los datos tabulares?

Los datos tabulares son aquellos que pueden organizarse en una estructura de tabla, es decir, un conjunto de:

* __Filas__, también llamadas observaciones o registros.
* __Columnas__, también llamadas variables.
* __Celdas__, cada celda debe contener un único valor.

__Nota.__ Ese único valor puede ser un conjunto de valores, es decir, una celda puede contener, por ejemplo, una lista.

Un ejemplo muy sencillo de dato tabular sería la siguiente tabla:

| | Animal      | Es mamífero| Peso_kg | Habitat       |
|-|-------------|------------|---------|---------------|
|1| León        | True       | 190     | Sabana        |
|2| Águila      | False      | 6       | Montaña       |
|3| Tiburón     | False      | 800     | Océano        |
|4| Rana        | False      | 0.5     | Humedal       |
|5| Elefante    | True       | 5400    | Sabana        |
|6| Pingüino    | False      | 30      | Antártida     |


Observamos que la tabla puede contener distintos tipos de dato como strings, booleanos o cifras.

Los datos tabulares se pueden representa como una matriz, pero entonces, si ya podíamos usar NumPy qué sentido tiene trabajar con Pandas.

Pandas parte de las representaciones de arrays de NumPy (está construido sobre NumPy de hecho) pero añade una semántica adicional al permitir:

* Indicar un identificador a las columnas (Animal, Es mamífero, Peso_kg y Habitat en nuestra tabla de ejemplo)
* Indicar un identificador a las filas (en el ejemplo un numérico pero podría ser un string)
* En columnas distintas podemos trabajar con tipos de datos distintos.

Las dos priemras características nos permiten trabajar de manera más cómoda y sencilla a la hora de procesar la información pero podrían llegar a suplirse con NumPy. Sin embargo, la combinación de distintos tipos de elementos conservando una alta eficiencia es un cambio en el paradigma de trabajo respecto a NumPy. Pasamos de trabajar con estructuras homogéneas (arrays) a estructuras heterogéneas (tablas en Pandas).

__Nota.__ Podemos usar tipos distintos en columnas distintas de manera que nuestra tabla contiene varios tipos pero no podemos usar tipos distintos dentro de una misma columna. Podemos pensar en cada columna como si se tratase de un array de NumPy.

### Estructuras de datos en Pandas

En Pandas distinguimos tres estructuras de dato clave:

* Series
* DataFrames
* Index

Profundizaremos en cada uno de ellos a lo largo de este curso pero de momento veremos una breve definición para cada uno:

__Def.__ Una __Series__ es un array unidimensional etiquetado, compuesto por un array de datos (basado en NumPy) y un índice.

Veamos un ejemplo simple:

```python
dias_mes = pd.Series(
    [31, 28, 31, 30, 31],
    index=["enero", "febrero", "marzo", "abril", "mayo"]
)

dias_mes
```

__Def.__ Un __DataFrame__ es una colección bidimensional de series alineadas por un índice común. Cada columna del DataFrame es una serie y todas las series comparten un índice. 

Es la estructura que abstrae el concepto de tabla con el que estamos acostumbrados a trabajar en otras herramientas como Excel.

Veamos cómo modelizar la tabla que vimos al principio de esta clase en Pandas:

```python
df_animales = pd.DataFrame(
    {
        "Animal": ["León", "Águila", "Tiburón", "Rana", "Elefante", "Pingüino"],
        "Es mamífero": [True, False, False, False, True, False],
        "Peso_kg": [190, 6, 800, 0.5, 5400, 30],
        "Habitat": ["Sabana", "Montaña", "Océano", "Humedal", "Sabana", "Antártida"]
    },
    index=[1, 2, 3, 4, 5, 6]  # índice personalizado
)

df_animales
```

Cada columna es una serie y todas comparten el índice numérico, es decir, todas van del 1 al 6. Ninguna serie podría tener un elemento 7 si todas las otras no lo tuvieran.

__Def.__ El objeto __Index__ es una estructura inmutable que representa las etiquetas de los ejes y permite que las estructuras se alineen de manera automática, es decir, mantener la información de manera coherente entre las distintas Series que componen un DataFrame.

Podemos observar los índices respectivos de las dos estructuras que hemos creado hasta ahora:

```python
dias_mes
```

```python
dias_mes.index
```

En el caso de nuestra serie, el índice está compuesto por los nombres de los distintos meses. Si ahora ordenáramos este índice, por ejemplo, alfabéticamente, los valores de los días se reordenarían igualmente para mantener la coherencia de la información.

```python
df_animales
```

```python
df_animales.index
```

En este otro caso observamos que el índice está compuesto por los enteros del 1 al 6. Además, si consultamos el índice asociado a cualquiera de las columnas veremos que es el mismo ya que __el índice se define para toda la tabla__:

```python
df_animales.Animal.index
```

```python
df_animales.Habitat.index
```

## 2.2. Series en Pandas

En esta clase vamos a:

- Explorar las Series en pandas entendiendo su estructura interna
- Comprender la relación que se genera entre índice y valor
- Estudiar sus atributos más importantes
- Comparar las Series con dos estructuras similares que hemos estudiado previamente: las listas y los arrays de NumPy.

Recordemos que una serie es una estructura de datos unidimensional etiquetada, basada en un array de NumPy.

### Estructura interna

Internamente una lista contiene:

* Un array de datos
* Un objeto Index
* Metadatos que nos aportan información sobre el contenido (tipo de objeto, tamaño...)



```python
import pandas as pd

dias_maxima = pd.Series(
    [20, 18, 22, 21, 16, 10, 5],
    index=["lunes", "martes", "miércoles", "jueves", "viernes",
          "sábado", "domingo"]
)

dias_maxima
```

`dias_maxima` tiene como array de datos un array de enteros que recoge la temperatura y como índice la lista de días de la semana. Sus metadatos incluirían cosas como el tamaño o el tipo de dato almacenados en el array.

Podemos pensar en las Series como un array con etiquetas o como un diccionario de Python ordenado y optimizado para operaciones.

Algunos de los usos más habituales en la práctica son utilizar las series como variables en una tabla, como series temporales, como vectores de resultados o como métricas agregadas.

### Relación índice valor

La relación entre índice y valor se mantiene de forma explícita dentro de las Series y nos permite acceder de manera rápida a un valor:

```python
dias_maxima["sábado"]
```

El índice es una estructura optimizada para búsqueda rápida que puede contener tanto strings como números o fechas. Además es una estructura inmutable. 

La clave del índice es que mantiene el valor asociado en todo momento de manera que si deseamos hacer otras operaciones, la asociación entre índice y valor se da de manera automática. Imaginemos que nos indican el incremento de temperaturas para la semana siguiente en otra serie y deseamos poder sacar la temperatura agregada:

```python
incr_temp = pd.Series(
    [-10, -8, -8, 10, 12, 14, 20],
    index=["domingo","lunes", "martes", "miércoles", "jueves", "viernes",
          "sábado"]
)
incr_temp
```

Si no tuviéramos los índices tendríamos que ver cómo reordenar nuestras series para que sigan el mismo orden. Sin embargo, gracias a la indexación podemos hacer simplemente:

```python
dias_maxima+incr_temp
```

Observamos cómo la suma ha ignorado las posiciones y ha asociado correctamente los valores basándose en los índices. Este es uno de los puntos clave del trabajo con Pandas y una de las mejoras significativas en comparación con NumPy.

Los atributos de una Serie nos permiten comprender mejor la información que se encuentra contenida en su interior. A continuación, veremos una lista de los más importantes:

```python
dias_maxima.dtype
```

* `.name` nos permite obtener o asignar un nombre a la serie. Es especialmente útil cuando en un futuro una serie formará parte de un DataFrame:

```python
dias_maxima.name
```

Vemos que la serie no tiene un nombre asignado (podría haberse asignado en su creación) así que se lo asignamos:

```python
dias_maxima.name = "temp_maxima"
```

```python
dias_maxima.name
```

* `.shape` nos permite conocer la forma de la serie. En este caso como siempre es unidimensional será siempre una tupla con un único valor que indica el número de elementos que la conforman:

* `size` nos permite conocer el número total de elementos que componen la serie:

```python
dias_maxima.size
```

* Por último, aunque no son atributos per se, los métodos `.head()`y `.tail()` son especialmente útiles cuando trabajamos con objetos de gran tamaño porque nos permiten ver una pequeña muestra de los mismos:

`head()` muestra los primeros elementos:

```python
dias_maxima.head()
```

`tail()` por el contrario muestra los últimos:

```python
dias_maxima.tail()
```

### Comparativa de Series con arrays de NumPy y listas

La siguiente tabla resume las principales diferencias entre una lista y una serie:

| Característica           | Lista       | Series    |
| ------------------------ | ----------- | --------- |
| Tipado                   | Heterogéneo | Homogéneo |
| Índice                   | Implícito   | Explícito |
| Operaciones vectorizadas | No          | Sí        |
| Alineación automática    | No          | Sí        |


Como hemos visto las series deben contener un único tipo:

```python
serie_het = pd.Series(
    [1, 2.3, True, "Hola", 16]
)

serie_het
```

La Serie fuerza los valores al tipo más general `object` para forzar la homogeneidad. Por el contrario una lista:

```python
lista_het = [1, 2.3, True, "Hola", 16]
lista_het
```

Además, podemos observar que en la serie el índice se encuentra de manera explícita frente a la lista que tiene un índice pero es implícito.

Además frente a las listas, las Series cuentan con operaciones vectorizadas y alineación automática que mantiene la cohesión en la información.

En cuanto a los arrays de NumPy:

| Característica           | NumPy Array | Series     |
| ------------------------ | ----------- | ---------- |
| Dimensionalidad          | 1D o nD     | 1D         |
| Índice                   | Posicional  | Etiquetado |
| Operaciones matemáticas  | Sí          | Sí         |
| Alineación por etiquetas | No          | Sí         |

Como hemos visto previamente a la hora de realizar operaciones Pandas añade la funcionalidad de alineamiento automático que nos ayuda a mantener el orden en las operaciones frente a la suma por posición habitual en NumPy:


```python
import numpy as np

a = np.array([10, 20])
b = np.array([1, 2])

a + b
```

```python
s1 = pd.Series([10, 20], index=["a", "b"])
s2 = pd.Series([1, 2], index=["b", "a"])

s1 + s2
```

### Tipos de datos en series

En Pandas, como hemos visto previamente, una serie tiene un único tipo subyacente que podemos extraer mediante el atributo `dtype`:

```python
import pandas as pd

serie_ejemplo = pd.Series([1, 2, 3])
print(serie_ejemplo)
print(serie_ejemplo.dtype)
```

La elección de tipo es capital ya que el tipo de la serie determina cómo se almacenan los datos en memoria, qué operaciones están disponibles y cómo se comportan los valores faltantes.

A nivel funcional, por ejemplo, si una columna numérica de una tabla con la que estamos trabajando se carga como un string no permitirá realizar operaciones como sumas o restas o de hacerlas requerirá más tiempo del esperado.

#### Tipos numéricos

Los tipos numéricos son todos aquellos heredados de NumPy y que se han explorado en profundidad en el curso anterior de Introducción a NumPy. Recorremos brevemente cada tipo viendo algún ejemplo:

**Enteros sin signo**

```python
serie_entero_sin_signo = pd.Series([1, 2, 3], dtype='uint8')
print(serie_entero_sin_signo)
print(serie_entero_sin_signo.dtype)
```

**Enteros**

```python
serie_entero_con_signo = pd.Series([1, -2, 3])
print(serie_entero_con_signo)
print(serie_entero_con_signo.dtype)
```

**Decimales (con coma flotante)**

```python
serie_decimales = pd.Series([1.5, -2/3, 3])
print(serie_decimales)
print(serie_decimales.dtype)
```

Podemos forzar un tipo de flotante concreto si quisiéramos reducir memoria:

```python
serie_decimales = pd.Series([1.5, -2/3, 3], dtype="float32")
print(serie_decimales)
print(serie_decimales.dtype)
```

**Complejos**

```python
serie_complejos = pd.Series([1, 2+3j, 4])
print(serie_complejos)
print(serie_complejos.dtype)
```

Como vimos previamente en NumPy, los tipos numéricos se adaptan entre ellos de manera automática para la realización de operaciones mediante reglas de promoción:

```python
serie_resultado = serie_entero_con_signo + serie_complejos
print(serie_resultado)
```

Además es interesante ver que para cualquier serie que contenga un valor `nan` el tipo se asignará automáticamente a float:

```python
pd.Series([1, 2])
```

```python
pd.Series([1, 2, None])
```

#### Tipo object

Es un tipo que nos permite almacenar objetos más generales; desde strings hasta mezcla de tipos u objetos Python:

```python
pd.Series(["Ana", True, 3])
```

Aunque permite una mayor versatilidad, `object` es lento comparado con otros tipos especializados y debemos evitarlo siempre que podamos.

#### Tipo string

Este tipo nos permite almacenar cadenas de caracteres. En versiones anteriores de Pandas estos se almacenaban en Series de tipo `object` pero este tipo fue creado para poder mejorar la eficiencia en el caso de trabajar con Series de cadenas de caracteres:

```python
serie_string = pd.Series(["Hola", "soy", "una serie"])
serie_string
```

#### Tipo booleano

Este tipo nos permite almacenar valores True o False y operar con ellos de manera eficiente:

```python
serie_bool = pd.Series([True, False, False])
serie_bool
```

Si nuestra serie contiene valores nulos esta será casteada automáticamente a `object`:

```python
serie_bool_nulo = pd.Series([True, False, False, None])
print(serie_bool_nulo)
```

#### Tipo fecha

En este tipo podemos almacenar datos tanto de fecha como de tiempo (es decir, horas minutos y segundos):

```python
serie_temp = pd.Series(["2024-01-01","2025-01-01","2026-01-01"], dtype="datetime64[ns]")
serie_temp
```

__Nota.__ En el último curso de esta especialidad se abordará en profundidad el trabajo y análisis de series temporales.

#### Tipo category

En ocasiones el dato con el que vamos a trabajar toma un conjunto cerrado de categorías, por ejemplo, estado civil iría recogido en  un string pero no cualquier string vale solo un conjunto finito de ellos:

```python
serie_cat = pd.Series(["Casado", "Soltero", "Separado", "Soltero"], dtype="category")
serie_cat
```

Este tipo de almacenamiento destaca por su menor consumo de memoria y por la mejora del rendimiento cuando se procede a agrupaciones y es un tipo muy usado en el análisis de datos.

#### Tipos extendidos

Además de todos los tipos anteriores, Pandas crea una serie de tipos extendidos que permiten añadir nulos a categorías que tradcionalmente no los tenían como los enteros y los booleanos a través de los tipos `Int64` y `boolean`:

```python
serie_entero_con_signo = pd.Series([1, -2, 3, None], dtype="Int64")
serie_entero_con_signo
```

```python
serie_bool_nulo_cast = pd.Series([None, False, False, True], dtype="boolean")
serie_bool_nulo_cast
```

Esta extensión de tipos es realmente util en Pandas ya que el trabajo con valores perdidos es bastante habitual y permite conservar muchas de las eficiencias de los tipos enteros y booleanos a nivel memoria que se perdían previamente a esta extensión de tipos.

### Integración de tipos

Al igual que en los arrays de NumPy las series no pueden contener más de un tipo por lo que en caso de haber mezcla, Pandas aplicará reglas de promoción para almacenar la mayor cantidad de información imposible de manera automática:

```python
enteros = pd.Series([1,2,3])
enteros_y_float = pd.Series([1,2.2,3])
print(enteros, "\n\n", enteros_y_float)
```

### Conversión de tipos

El método más habitual para la conversión de tipos es `.astype()` que nos permite indicar de manera explícita a que tipo deseamos convertir nuestra serie:

```python
enteros
```

```python
enteros_a_float = enteros.astype(float)
enteros_a_float
```

```python
enteros_a_string = enteros.astype(str)
enteros_a_string
```

Además, Pandas cuenta con dos métodos específicos; `pd.to_numeric()` y `pd.to_datetime()` que permiten respectivamente convertir de manera segura los tipos:

```python
serie_mixta = pd.Series([1, 2, "Hola", True, 4])
serie_mixta
```

```python
serie_numerica = pd.to_numeric(serie_mixta)
serie_numerica
```

El error nos indica que el método no es capaz de transformar el string "Hola" a un número. El método incluye el argumento `errors` que nos permite indicar que use nulos cuando no sea capaz de realizar la conversión:

```python
serie_numerica_nan = pd.to_numeric(serie_mixta, errors="coerce")
serie_numerica_nan
```

Análogamente para fechas:

```python
serie_temp_mixta = pd.Series(["2024-01-01","2025-01-01","2026-01-01", "Hola"])
serie_temp_mixta
```

```python
serie_fechas = pd.to_datetime(serie_temp_mixta)
serie_fechas
```

```python
serie_fechas = pd.to_datetime(serie_temp_mixta, errors="coerce")
serie_fechas
```

### Impacto en memoria

Es interesante cerrar esta clase viendo cómo la elección de tipado impacta significativamente el uso de memoria, por ejemplos:

```python
serie_str = pd.Series(["Madrid", "Oviedo", "Oviedo", "Oviedo", "Madrid", "Madrid", "Madrid"])
serie_str
```

```python
serie_cat = pd.Series(["Madrid", "Oviedo", "Oviedo", "Oviedo", "Madrid", "Madrid", "Madrid"], dtype="category")
serie_cat
```

```python
print(serie_str.memory_usage(), serie_cat.memory_usage())
```

Observamos como ya en un ejemplo muy pequeño la diferencia es bastante significativa. Como sabemos, uno de los principales usos de Pandas es el trabajo con grandes cantidades de datos por lo que imaginemos lo significativa que sería esta diferencia si trabajaramos con series que contuvieran millones de elementos.

Es una muy buena práctica siempre que trabajemos con una serie validar su tipo y comprobar si tiene sentido castearla a algún tipo en aras de mejorar la eficiencia y el uso de recursos.

## 2.4. Creación de Series en Pandas

Hasta ahora hemos creado las Series de manera manual a partir de distintas listas. En esta clase exploraremos otras formas de crear series, por ejemplo a partir de diccionarios o arrays de NumPy. Además, exploraremos la creación de series mediante valores aleatorios que nos servirán para crear ejemplos para este curso análogamente como hicimos con NumPy en el curso anterior.

El método más sencillo con el que hemos trabajado hasta ahora es `pd.Series()` que nos permite crear una Serie a partir de una lista de datos e incluso dar algo de información al respecto de la misma:

```python
import pandas as pd

serie_ejemplo = pd.Series(data=[1,2,3,4,5,6,7],
                          index= ["EN", "FE", "MA", "AB", "MY", "JN", "JL"],
                          dtype=float,
                          name="meses"
                         )

serie_ejemplo
```

Como hemos visto previamente, si no indicamos un índice de manera explícita, Pandas genera un índice de manera automática:

```python
import pandas as pd

serie_ejemplo_sin_ind = pd.Series(data=[1,2,3,4,5,6,7],
                          dtype=float,
                          name="meses"
                         )

serie_ejemplo_sin_ind
```

Este método es muy práctico para la creación de ejemplos y para iniciarse en el mundo de Pandas pero resulta bastante rudimentario y poco práctico cuando se desea trabajar con series de gran tamaño.

### Creación a partir de un array

Dado que Pandas está construido sobre NumPy la realización de esta conversión es muy sencilla:

```python
import numpy as np

array = np.array([100, 200, 300])
serie_array = pd.Series(array)
serie_array
```

La serie infiere el dtype de manera directa desde NumPy. Si lo deseáramos podríamos añadirle un índice a esta serie:

```python
serie_array_ind = pd.Series(array, index=["A", "B", "C"])
serie_array_ind
```

Este método puede resultar especialmente útil cuando queremos integrar en Series de Pandas resultados de un análisis previo construido en NumPy.

#### Creación a partir de un diccionario

Este método nos permite transformar los pares clave-valor que conforman un diccionario, en los pares indice-valor que conforman una serie:

```python
datos = {
    "a": 10,
    "b": 20,
    "c": 30
}

serie_diccionario = pd.Series(datos)
serie_diccionario
```

Además es interesante observar qué ocurre si indicamos un índice explícito que no encaja con el índice generado por las claves del diccionario:

```python
datos = {
    "a": 10,
    "b": 20,
    "c": 30
}

serie_diccionario_ind = pd.Series(datos, index=["a", "b", "d"])
serie_diccionario_ind
```

Esto muestra dos conceptos fundamentales:

## 2.6. Acceso a elementos en series

En esta clase vamos a explorar las distintas opciones con las que contamos para poder acceder a elementos dentro de una serie, así como métodos para modificar los valores a los que accedemos.

### Acceso por posición y acceso por etiqueta

La manera más intuitiva si venimos de trabajar con Python a la hora de acceder a un elemento es el acceso mediante posición, análogo al de las listas:

```python
import pandas as pd

serie_ejemplo_acceso = pd.Series(np.random.randint(100,200,10))
serie_ejemplo_acceso
```

```python
serie_ejemplo_acceso[0]
```

```python
serie_ejemplo_acceso[5]
```

Aparentemente estamos accediendo a la posición indicada. Esta idea es un error bastante común al empezar a trabajar con series en Pandas. Realmente lo que estamos haciendo es acceder a la etiqueta que en este caso coincide con la posición. Veamos qué ocurre cuando lo hacemos con una serie que no tenga un índice numérico:

```python
serie_ejemplo_acceso_2 = pd.Series(np.random.randint(100,200,10),
                                 index = ["A", "B", "C", "D", "E",
                                          "F", "G", "H", "I", "J"]
                                )
serie_ejemplo_acceso_2
```

```python
serie_ejemplo_acceso_2[0]
```

En este caso nos arroja un error de clave y nos indica que no es capaz de encontrar la etiqueta 0. Sin embargo, si usamos una etiqueta definida el acceso se produce sin problemas:

```python
serie_ejemplo_acceso_2["C"]
```

Para evitar estas confusiones una buena práctica es utilizar los métodos `.loc` e `.iloc`.

### Métodos `.loc` e `.iloc`.

El método `.loc` trabaja únicamente con etiquetas de la serie lo cual nos garantiza un acceso estable ya sea mediante índices numéricos o un índice de string:

```python
serie_ejemplo_acceso.loc[3]
```

```python
serie_ejemplo_acceso_2.loc["D"]
```

Si en el segundo caso intentáramos usarlo con un índice numérico de posición nos arrojaría un error:

```python
serie_ejemplo_acceso_2.loc[5]
```

__Nota.__ Es importante tener en cuenta que el índice de una serie no tiene por qué ser único lo que provocaría que en caso de usar el método `.loc` sobre un índice repetido obtuviéramos una serie con más de un valor:

```python
ejemplo_indice_repetido = pd.Series([1,2,3,4], index=["A", "B", "A", "C"])
ejemplo_indice_repetido
```

```python
ejemplo_indice_repetido.loc["A"]
```

Sin embargo, en ocasiones puede ser útil acceder a un elemento por su posición y no por su etiqueta. Supongamos por ejemplo, que hemos sumado todos los consumos de nuestros clientes y queremos acceder a los cinco clientes más provechosos cuyo nombre no conocemos de antemano. En este caso necesitaríamos poder acceder a los elementos mediante su posición y para eso contamos con el método `.iloc`. En el caso de la primera serie que ya usaba un índice numérico el resultado es el mismo que con `.loc`:

```python
serie_ejemplo_acceso.loc[3]
```

```python
serie_ejemplo_acceso.iloc[3]
```

Sin embargo, en el caso de nuestra segunda serie ahora sí podremos acceder por posición:

```python
serie_ejemplo_acceso_2.iloc[5]
```

### Slicing

En ocasiones vamos a querer acceder a más de un elemento a la vez. Pandas extiende la noción de slicing ya implementada en NumPy para este tipo de necesidades.

Con el método `.iloc` el slicing funciona como cualquier otro slicing en Python: 

```python
serie_ejemplo_acceso.iloc[2:5]
```

```python
serie_ejemplo_acceso_2.iloc[2:5]
```

__Nota.__ Recordemos que se excluye siempre el extremo superior del intervalo

Con el método `.loc`, sin embargo, el slicing se comporta un poco diferente pues __incluye__ el extremo superior del intervalo:

```python
serie_ejemplo_acceso.loc[1:4]
```

```python
serie_ejemplo_acceso_2.loc["C":"I"]
```

__Nota.__ Vemos como tanto la posición 4 como la posición I se encuentran incluidas.

### Acceso a múltiples elementos

En ocasiones los distintos elementos a los que queremos acceder no se van a encontrar seguidos por lo que no podremos valernos del slicing. En estos casos utilizaremos el `fancy indexing` heredado igualmente de NumPy que es compatible tanto con `.loc` como con `.iloc`:

```python
serie_ejemplo_acceso.loc[[1,4]]
```

```python
serie_ejemplo_acceso_2.loc[["A","F","I"]]
```

```python
serie_ejemplo_acceso_2.iloc[[0,5,8]]
```

### Acceso mediante condiciones

Las Series de Pandas heredan el concepto de máscara de NumPy, un array que se genera a partir de la evaluación de una condición sobre todos los elementos que componen un array. En este caso la condición se validará sobre todos los elementos que componen una serie:

## 2.7. Trabajando con valores nulos

En esta clase vamos a entender exactamente qué es un valor nulo o vacío viendo no solo nociones técnicas sobre su generación y funcionamiento sino también desde un enfoque conceptual, entendiendo para qué sirve o qué suelen representar.

Un valor nulo representa __ausencia de dato__. Es importante entender que no es cero, no es una cadena vacío y no es un valor False, es un marcador especial que indica que el valor del dato es desconocido o no está disponible.

En Pandas suele aparecer como:

* `np.nan`
* `None`
* `pd.NA`

dependiendo del `dtype` de la serie en que se encuentre.

Conceptualmente es esencial comprender que un nulo nos da información sobre la ausencia de valor. Esto es más fácil de entender con algunos ejemplos prácticos:

- Un cliente que no responde una pregunta en una encuesta.
- Un sensor que falla durante una medición.
- Un dato que aún no existe.
- Un error en la carga de datos.
- Un campo opcional que no se rellena en un formulario.

Para crear una Serie con un valor pérdido podemos usar simplemente:




```python
import numpy as np
import pandas as pd

serie_con_nulo = pd.Series([1, 2, None, 3, 10, None])
serie_con_nulo
```

__Nota.__ Cuando introducimos un nulo en una serie numérica esta se convierte automáticamente a tipo float.

En las últimas versiones aparece `pd.NA` que funciona de manera compatible con enteros y booleanos "nullables":

```python
otra_serie_con_nulo = pd.Series([1, 2, pd.NA, 3])
otra_serie_con_nulo
```

```python
serie_entero_nullable = pd.Series([1, pd.NA, 3], dtype="Int64")
serie_entero_nullable
```

### Detección de valores nulos

Nunca debemos utilizar la comparativa directa para detectar nulos como haríamos con otros valores:

```python
serie_con_nulo == np.nan
```

Al usarse los NaN para valores indeterminados Python no puede determinar si dos valores indeterminados son el mismo valor por lo que __siempre__: `NaN != NaN`. 

Para detectar valores nulos tenemos dos métodos específicos: `.isnull()` y `.isna()`:

```python
serie_con_nulo.isna()
```

```python
serie_con_nulo.isnull()
```

Para realizar una tarea habitual como es contar el número de valores nulos dentro de una serie combinamos el método anterior con el método `.sum()` y las reglas de promoción que transformarán los valores `True` en 1 y los `False` en 0:

```python
serie_con_nulo.isna().sum()
```

Otras operaciones bastante habituales consisten en comprobar si existe algún valor nulo o si todos lo son:

```python
serie_con_nulo.isna().any()
```

```python
serie_con_nulo.isna().all()
```

### Nulos y operaciones

Como premisa básica `Pandas` implementa la propagación de nulos que implica que cualquier operación que implica nulos devuelve valores nulos:

```python
serie_con_nulo + 2
```

Esta idea es bastante razonable ya que si no conocemos uno de los elementos de la opración muy difícilmente podríamos conocer su resultado salvo en casos muy concretos.

Existen algunas operaciones que por defecto no tienen en cuenta los valores nulos porque entienden que desvirtuarían el resultado, por ejemplo:

```python
serie_con_nulo.mean()
```

Estas operaciones se han codificado con el argumento `skipna` a True por defecto pero se puede modificar de manera sencilla:

```python
serie_con_nulo.mean(skipna=False)
```

### Tratamiento de valores nulos

A la hora de afrontar el tratamiento de valores nulos existen multitud de opciones según el caso ante el que nos encontremos. A continuación recorreremos los tratamientos más habituales.

#### Eliminar valores nulos

En ocasiones la proporción de valores nulos es ínfima y realmente no están aportando información para el problema que queremos resolver. Supongamos, por ejemplo, que tenemos las mediciones de un termómetro durante todos los minutos del año. Por algún motivo este termómetro arroja algunos valores nulos aislados que suponen un 0.01% de las mediciones y que no siguen un patrón especial.

Si nuestro objetivo es estudiar la temperatura durante el año, podemos prescindir de los valores nulos y estudiar todos los demás:

```python
serie_sin_nulo = serie_con_nulo.dropna()
serie_sin_nulo
```

```python
serie_con_nulo
```

Hemos prescincido del par índice-valor que tenía un valor nulo y conservado todo lo anterior de manera idéntica.

__Nota.__ Es importante ver que estos métodos reducen el tamaño de la serie de datos con la que trabajamos:

```python
print(serie_sin_nulo.size, serie_con_nulo.size)
```

#### Rellenar valores nulos

En ocasiones, puede que prefiramos conservar esos valores, por ejemplo, si suponen un tamaño relevante de nuestros datos. Pandas pone a nuestra disposición el método `fillna()` que nos permite sustituir los valores nulos por otros valores. La clave de este método ya no reside en la dificultad técnica sino funcional de cómo determinar ese valor.

Una idea inicial podría ser sustituir los valores nulos por un mismo valor, por ejemplo, si tengo una serie que recoge la respuesta a número de hijos en un formulario, puedo teorizar que la gente que no responde es porque no los tiene y sustituir por cero:

```python
numero_hijos =  pd.Series(np.random.randint(1,5,10))
numero_hijos[4] = None
numero_hijos[7] = None
numero_hijos
```

```python
numero_hijos_imputado = numero_hijos.fillna(0)
numero_hijos_imputado
```

Si volviéramos al caso anterior de las temperaturas pero en lugar de un 0,01% de nulos tuviéramos un 10% y una vez analizado que no siguen un patrón claro podríamos plantearnos la imputación pero... en este caso debemos pensar con algo más de cabeza con qué número imputar. Las opciones más habituales son:

```python
serie_imputa_media = serie_con_nulo.fillna(serie_con_nulo.mean())
serie_imputa_media
```

```python
serie_imputa_mediana = serie_con_nulo.fillna(serie_con_nulo.median())
serie_imputa_mediana
```

Por último en este mismo ejemplo de las temperaturas podría tener sentido decidir que conservaremos el valor que se encuentre en el minuto siguiente o en el minuto anterior ya que podemos asumir que no existen variaciones muy significativas en la temperatura de un minuto a otro:

```python
serie_imputa_pre = serie_con_nulo.bfill()
serie_imputa_pre
```

```python
serie_imputa_post = serie_con_nulo.ffill()
serie_imputa_post
```

La elección de la estrategia para el tratamiento de valores nulos suele ser una de las claves para una interpretación exitosa de los datos. Esta elección requiere combinar conocimiento técnico y funcional y en muchos casos puede que tengamos que cotejar resultados con distintos métodos hasta tomar una decisión definitiva o incluso realizar tratamientos distintos a los mismos datos según el objetivo que estemos persiguiendo.

```python
serie_ejemplo.max()
```

Calcula la desviación estándar de una serie:

```python
serie_ejemplo.std()
```

Calcular el valor del cuantil indicado en la distribución definida por la serie:

```python
percentil_90 = serie_ejemplo.quantile(0.9)
print(percentil_90)
```

#### `.describe()`

Es especialmente útil cuando nos enfrentamos a unos datos por primera vez porque nos permite con un solo método obtener una visión global de los primeros estadísticos:

#### `.count()`

Permite contar los valores no nulos:

```python
serie_ejemplo.count()
```
