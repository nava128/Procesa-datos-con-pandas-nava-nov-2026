# Sección 5. Selección y filtrado de datos

En esta sección vamos a explorar los distintos métodos de selección y filtrado de datos que nos permiten partiendo de un DataFrame completo extraer solo los registros que cumplen determinadas condiciones en sus variables.

## 5.1. Subconjuntos de datos

En esta clase vamos a explorar cómo extraer subconjuntos de datos de un DataFrame ya sea seleccionando columnas o filas específicas.

__Def.__ Un __subconjunto de datos__ es una parta del DataFrame original obtenida mediante algún criterio de selección.

Puede consistir en:

* Una o varias columnas.
* Una o varias filas
* Una combinación de filas y columnas

Por ejemplo, dado el siguiente DataFrame:

|   | nombre | edad | ciudad   | salario |
| - | ------ | ---- | -------- | ------- |
| 0 | Ana    | 25   | Oviedo   | 3500    |
| 1 | Diana  | 30   | Sevilla  | 2000    |
| 2 | Carlos | 22   | Madrid   | 1200    |


Un ejemplo de subconjunto sería:

| nombre | salario |
| ------ | ------- |
| Ana    | 3500    |
| Carlos | 1200    |

### Selección de filas y columnas

Para seleccionar una o varias columnas la forma más sencilla de hacerlo es utilizando su nombre:

```python
import pandas as pd

data = {
    "nombre": ["Ana", "Diana", "Carlos"],
    "edad": [25, 30, 22],
    "ciudad": ["Oviedo", "Sevilla", "Madrid"],
    "salario": [3500, 2000, 1200]
}

df_seleccion = pd.DataFrame(data)
df_seleccion
```

```python
df_seleccion["ciudad"]
```

__Nota.__ Mediante el corchete simple extraemos la Serie que compone la columna.  Si usamos corchete doble extraemos un DataFrame formado únicamente por esa columna:

```python
df_seleccion[["ciudad"]]
```

```python
df_seleccion[["nombre", "ciudad"]]
```

### Métodos `.loc` e `.iloc`

#### `.loc`


Para seleccionar una o varias filas podemos utilizar los métodos `.loc` e `.iloc` que hemos estudiado en clases previas sobre Series. Vamos a añadir un índice explícito para dejar clara la diferencia entre estos dos métodos:

```python
df_seleccion.index = ["A","B","C"]
df_seleccion
```

`.loc` permite seleccionar datos utilizando las etiquetas del índice siguiendo la sintaxis: `.loc[filas,columnas]`.

Para seleccionar una sola fila:

```python
df_seleccion.loc["A"]
```

De nuevo corchete simple para extraer una Serie y corchete doble para extraer un DataFrame de una sola fila:

```python
df_seleccion.loc[["A"]]
```

Para seleccionar varias filas:

```python
df_seleccion.loc[["A","C"]]
```

Para seccionar un rango de filas:

```python
df_seleccion.loc["A":"C"]
```

Veamos algunos ejemplos combinando extracción de filas y columnas:

```python
df_seleccion.loc["A", "edad"]
```

```python
df_seleccion.loc[["A","C"], "edad"]
```

```python
df_seleccion.loc[["A","C"], ["nombre","edad", "ciudad"]]
```

#### `.iloc`

Si en lugar de acceder a los elementos por su etiqueta deseamos acceder por su posición contamos con el método de funcionamiento análogo `.iloc[]`:

```python
df_seleccion.iloc[2]
```

```python
df_seleccion.iloc[[2]]
```

```python
df_seleccion.iloc[[0,2]]
```

```python
df_seleccion.iloc[0:3]
```

Al igual que `.loc`, `.iloc` nos permite acceder a subconjuntos que combinen filas y columnas:

```python
df_seleccion.iloc[[0,2],1:4]
```

La diferencia clave que debemos entender es que:

| Método  | Selección    |
| ------- | ------------ |
| `.loc`  | por etiqueta |
| `.iloc` | por posición |

__Nota.__ Recordemos que la indexación en Python empieza en 0, ergo, para acceder a la primera fila/columna mediante posición numérica usaremos el 0.

## 5.2. Selección condicional de datos en DataFrames

La selección condicional nos permite extraer los subconjuntos de datos de un DataFrame que cumplen unos determinados criterios.

__Def.__ La __selección condicional__ consiste en extraer filas de un DataFrame que cumplan una condición lógica. Este proceso se realiza utilizando __máscaras booleanas__ como ya hemos estudiado en el curso anterior sobre NumPy.

Recordemos que una máscara booleana es una estructura que contiene valores `True` y `False` que indican en cada caso si el valor en el objeto original cumple o no la condición evaluada.

Por ejemplo si tenemos el siguiente DataFrame:

| nombre | edad |
| ------ | ---- |
| Ana    | 25   |
| Luis   | 23   |
| Paula  | 22   |

Y generáramos una máscara según la condición `edad > 23` obtendríamos:

`True`

`False`

`False`

__Nota.__ Si has trabajado antes con SQL este sería el equivalente a una cláusula **WHERE**.

Partiendo del siguiente DataFrame veamos cómo podemos crear una máscara booleana y cómo utilizarla posteriormente para un filtrado:

```python
import pandas as pd

data = {
    "Origen": ["Madrid", "Sevilla", "Valencia", "Bilbao", "Zaragoza"],
    "Destino": ["Barcelona", "Madrid", "Sevilla", "Valencia", "Bilbao"],
    "Transporte": ["Avión", "Tren", "Bus", "Coche", "Avión"],
    "Precio": [120, 60, 30, 45, 100]
}

df_transporte = pd.DataFrame(data, index=["A", "B", "C", "D", "E"])
df_transporte
```

Para generar la máscara booleana simplemente debemos establecer nuestra condición y Pandas se ocupará de extenderla por el ámbito en el que la hayamos definido. Por ejemplo, creemos una máscara para la columna Transporte:

```python
masc_trans_columna = df_transporte["Transporte"] =="Avión"
masc_trans_columna
```

Si deseamos aplicar el filtrado al DataFrame de manera que conservemos solo las filas que refieren información sobre viajes en avión:

```python
df_transporte_avion = df_transporte[masc_trans_columna]
df_transporte_avion
```

Por último se puede definir la máscara sobre el DataFrame completo:

```python
masc_completa = df_transporte =="Valencia"
masc_completa
```

Esto puede ser útil por ejemplo, si queremos calcular cuántos trayectos tienen por destino o por origen Valencia:

```python
masc_completa.sum()
```

```python
df_transporte[masc_completa]
```

Normalmente para optimizar memoria y comprimir el código no se suele definir la máscara de manera explícita sino que se aplica en el mismo momento en el que se crea:

```python
df_transporte[df_transporte.Precio<100]
```

Recordemos a título informativo los distintos operadores de comparación definidos en Pandas:

| Operador | Significado   |
| -------- | ------------- |
| `==`     | igual         |
| `!=`     | distinto      |
| `>`      | mayor         |
| `<`      | menor         |
| `>=`     | mayor o igual |
| `<=`     | menor o igual |


Cerramos esta clase viendo que utilizando el operador `.loc` podemos igualmente aplicar máscaras:

```python
df_transporte.loc[df_transporte.Precio<100]
```

El resultado es exactamente idéntico al que obteníamos aplicando la máscara directamente usando corchetes.

La ventaja es que esto nos permite realizar a posteriori del filtrado una selección de columnas:

```python
df_transporte.loc[df_transporte.Precio<100, ["Origen","Destino"]]
```

Esto tiene sentido ya que si lo pensamos la máscara simplemente define una serie de filas a seleccionar.

## 5.3. Filtros avanzados en DataFrames

En esta clase retomaremos la idea de las máscaras pero veremos cómo combinarlas con otros operadores y métodos para poder obtener filtrados mucho más sofisticados.

### Uso de operadores lógicos

Pandas nos propone tres operadores lógicos básicos que nos permiten enlazar varias condiciones:

| Operador | Significado |
| -------- | ----------- | 
| `&`      | Y           | 
|  `\|`    |  O          |
| `~`      | NO          |



`&` nos permite combinar dos condiciones al mismo tiempo. Veamos el ejemplo sobre un DataFrame:

```python
import pandas as pd

data = {
    "Origen": ["Madrid", "Sevilla", "Valencia", "Bilbao", "Zaragoza"],
    "Destino": ["Barcelona", "Madrid", "Sevilla", "Valencia", "Bilbao"],
    "Transporte": ["Avión", "Tren", "Bus", "Coche", "Avión"],
    "Precio": [120, 60, 30, 45, 100]
}

df_transporte = pd.DataFrame(data, index=["A", "B", "C", "D", "E"])
df_transporte
```

Seleccionemos los trayector por encima de 35€ y con un máximo de 100€:

```python
df_transporte_35_100 = df_transporte[(df_transporte["Precio"] >35) & (df_transporte["Precio"] <=100)]
df_transporte_35_100
```

`|` nos permite extraer los registros que cumplan __al menos una__ de las dos condiciones. Podemos extraer por ejemplo los trayectos que tienen como origen o destino Sevilla:

```python
df_transporte_sev = df_transporte[(df_transporte["Origen"] =="Sevilla") | (df_transporte["Destino"] =="Sevilla")]
df_transporte_sev
```

`~` nos permite calcular la negativa de una condición. Por ejemplo, si quisiéramos todos los trayectos excepto aquellos que tienen origen o destino en Sevilla podríamos reutilizar la cláusula anterior y negarla:

```python
df_transporte_no_sev = df_transporte[~((df_transporte["Origen"] =="Sevilla") | (df_transporte["Destino"] =="Sevilla"))]
df_transporte_no_sev
```

### Métodos útiles para el filtrado avanzado

Pandas cuenta además con algunos métodos que suelen resultar útiles para implementar filtros avanzados. El funcionamiento de estos métodos suele poder reemplazarse por la concatenación de filtros y operadores lógicas pero el uso de estos métodos facilita la escritura y lectura del código.

El método `.isin()` comprueba si un valor pertenece a una lista de valores. 

Por ejemplo, si queremos comprobar qué trayectos tienen origen en Madrid, Bilbao o Zaragoza podemos comprimir la línea:

```python
df_transporte_bilb_mad_zgz = df_transporte[(df_transporte["Origen"]=="Bilbao") | (df_transporte["Origen"]=="Madrid") | (df_transporte["Origen"]=="Zaragoza")]
df_transporte_bilb_mad_zgz
```

En simplemente:

```python
df_transporte_bilb_mad_zgz = df_transporte[df_transporte["Origen"].isin(["Bilbao", "Madrid", "Zaragoza"])]
df_transporte_bilb_mad_zgz
```

El método `.between()` permite seleccionar valores dentro de un rango. Previamente hemos visto que para seleccionar trayectos entre 35 y 100:

```python
df_transporte_35_100 = df_transporte[(df_transporte["Precio"] >35) & (df_transporte["Precio"] <=100)]
df_transporte_35_100
```

Esto se puede expresar de manera mucho más concisa como:

```python
df_transporte_35_100 = df_transporte[df_transporte["Precio"].between(35,100)]
df_transporte_35_100
```

### Filtrado de strings

Pandas permite reutilizar muchos de los métodos típicos de la manipulación de strings dentro de los filtrados. Por ejemplo, podemos filtrar todos los medios de transporte que contengan una n:

```python
df_transporte_n = df_transporte[df_transporte["Transporte"].str.contains("n")]
df_transporte_n
```

O los de las ciudades que comienzan o terminan por una determinada letra:

```python
df_transporte_m_fin = df_transporte[df_transporte["Origen"].str.startswith("M")]
df_transporte_m_fin
```

```python
df_transporte_a = df_transporte[df_transporte["Origen"].str.endswith("a")]
df_transporte_a
```

__Nota.__ Es importante entender la potencia que esto implica cuando trabajamos con datasets de gran tamaño (decenas de miles de filas) y alguna columna por ejemplo, identifica según ciertos códigos unos productos de manera que por ejemplo, si queremos extraer todo lo que pertenece a la sección de tecnología, el código de producto simepre empieza por "tcno".

En ocasiones y cuando trabajemos con condiciones muy complejas o que podemos querer reutilizar en un futuro puede tener sentido almacenarlas y utilizarlas a posteriori para mejorar la legibilidad:

```python
origen_acaba_a = df_transporte["Origen"].str.endswith("a")
coste_inferior_100 = df_transporte["Precio"] < 100
```

```python
df_transporte[origen_acaba_a & coste_inferior_100]
```

Cerramos esta sección viendo el método `.query()` implementado en las últimas versiones de Pandas que nos permite utilizar una sintaxis más limpia y facilita la lectura e implementación de filtros complejos.

Veamos un ejemplo:

```python
df_transporte.query("Origen == 'Sevilla' or Precio > 100")
```
