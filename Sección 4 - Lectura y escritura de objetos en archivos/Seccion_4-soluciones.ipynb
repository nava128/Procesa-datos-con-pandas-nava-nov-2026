# Sección 4. Lectura y escritura de objetos en archivos

En esta sección vamos a estudiar cómo persistir y extraer datos en archivos. 

## 4.1. Almacenamiento de Series y DataFrames en archivos de datos

Los objetos Series y DataFrame existen en memoria mientras el programa está en ejecución. Pero qué ocurre si queremos reutilizar datos, almacenar resultados o exportar las informaciones... En ese caso debemos persistir los datos en un archivo para poder reutilizarlo más tarde o compartirlo con otros.

Pandas incluye múltiples métodos para escribir datos en diferentes formatos.

Existen multitud de formatos en los que podemos almacenar la información pero en esta clase nos centraremos en los más utilizados dentro de la industria y la academia:

| Formato | Ventaja                 | Desventaja         |
| ------- | ----------------------- | ------------------ |
| **CSV**     | simple y universal      | no conserva tipos  |
| **Excel**   | compatible con usuarios | más pesado         |
| **JSON**    | ideal para APIs         | menos legible      |
| **Pickle**  | muy rápido              | solo Python        |

### CSV

El archivo CSV (comma-separated-values) es uno de los más utilizados por su simplicidad y universalidad. Es un formato compatible con todos los programas de procesamiento de hojas de cálculos y con la mayor parte de bases de datos.

Para almacenar una serie o un DataFrame en este formato contamos con el método `.to_csv()`:


```python
import numpy as np
import pandas as pd

serie_guardar = pd.Series(np.random.randint(1,10,12), 
                          index=["A","B","C","D","E","F",
                                 "G","H","I","J","K","L"], 
                          name="Serie guardar"
                         )
serie_guardar
```

__Nota.__ Creamos una carpeta `data` donde almacenaremos nuestros resultados.

```python
!mkdir data
```

Para guardar esta serie en un archivo csv simplemente:

```python
serie_guardar.to_csv("./data/serie.csv")
```

Si observamos el archivo vemos que hemos almacenado el nombre de la serie seguido de todos los pares índide-valor. Además si lo abro con un gestor de hojas de cálculo veo que los formatos son compatibles.

Podemos hacer lo análogo con un DataFrame:

```python
data = {
    "nombre": ["Ana", "Paula", "Carlos"],
    "edad": [25.4, 30.2, 22.0],
    "ciudad": ["Madrid", "Sevilla", "Valencia"]
}

df_guardar = pd.DataFrame(data)

df_guardar.to_csv("./data/df_basic.csv")
df_guardar
```

De nuevo podemos observar cómo se guardan todos los pares índice valores junto con el nombre de las columnas.

El método `.to_csv()` es con diferencia el más utilizado en la manipulación de datos por lo que merece la pena detenerse algo más en su funcionamiento y sus argumentos.

Una de las funcionalidades interesantes es que podemos elegir si conservamos el índice o no. En nuestro caso por ejemplo, el índice no aporta ningún tipo de información por lo que podríamos prescindir de él para no ocupar espacio innecesario y optimizar los tiempos de carga y escritura así como la memoria en disco. Para ello podemos utilizar el argumento `index=False`:

```python
df_guardar.to_csv("./data/df_sin_ind.csv", index=False)
```

Al contrario que en el sistema estadounidense, en los países de habla hispana usamos la coma como indicador del decimal, es decir, diez y medio sería 10,5 y no 10.5. 

Por ello es muy habitual que para evitar confusiones especialmente si se planea la manipulación con procesadores de hojas de cálculo se utilice como separador el punto y coma (;) en lugar de la coma (,). 

Además, podemos indicar que queremos usar la "," como separador decimal:

```python
df_guardar.to_csv("./data/df_sin_ind_sep.csv", index=False, sep=";", decimal=",")
```

Igual que hemos podido prescindir del índice en ocasiones también querremos prescindir de los nombres de las columnas (aunque esto es bastante menos habitual). Para desecharlos usamos `header`:

```python
df_guardar.to_csv("./data/df_sin_col.csv", index=False, sep=";", decimal=",", header=False)
```

En ocasiones, especialmente cuando trabajemos con archivos con grandes cantidades de texto puede ser útil almacenarlos garantizando la codificación de caracteres para caracteres especiales como la ñ, las tildes... Para ello contamos con el argumento `encoding`:

```python
df_guardar.to_csv("./data/df_utf.csv", encoding="utf-8")
```

Por último, en ocasiones, a lo largo de un análisis hemos generado columnas auxiliares que no resultan útiles. `.to_csv()` nos permite indicar un listado de columnas mediante el argumento `columns` que serán las únicas que se guardarán:

```python
df_guardar.to_csv("./data/nombre_edad.csv", sep=";", decimal=",", columns=["nombre", "edad"])
```

### Excel

Excel es uno de los procesadores de hojas de cálculo más populares. ### JSON

Los archivos JSON son especialmente habituales cuando nuestros datos van a ser consumidos por APIs o aplicaciones web. 

El método `.to_json()` nos permite almacenar nuestros datos en un archivo de estas características de manera muy sencilla.

```python
serie_guardar.to_json("./data/serie.json")
df_guardar.to_json("./data/datos.json")
```

En esta línea los archivos `.html` pueden ser útiles si nuestro dato va a ir inyectado en una página web:

```python
df_guardar.to_html("./data/tabla.html")
```

### Pickle

Pickle convierte un objeto de Python en una secuencia de bytes (serialización) para poder guardarlo en un archivo, enviarlo por red o almacenarlo temporalmente.

Después se puede reconstruir el objeto original (deserialización).

```python
serie_guardar.to_pickle("./data/serie.pkl")
df_guardar.to_pickle("./data/datos.pkl")
```

__Nota importante.__ Los archivos pickle solo son seguros si provienen de fuentes confiables, porque al cargarlos pueden ejecutar código malicioso.

## 4.2. Lectura de Series y DataFrames a partir de archivos

En la clase anterior hemos visto los distintos métodos en los que podemos almacenar la información con la que trabajamos en Pandas dentro de archivos.

En esta clase veremos la operación recíproca. En la mayor parte de las ocasiones recibiremos los datos de un archivo o una base de datos y seremos nosotros los que deberemos encargarnos de cargar toda la información en Pandas para poder realizar posteriormente las manipulaciones y análisis pertinentes.

A lo largo de la clase exploraremos los siguientes métodos:

| Formato | Método              |
| ------- | ------------------- |
| CSV     | `pd.read_csv()`     |
| Excel   | `pd.read_excel()`   |
| JSON    | `pd.read_json()`    |
| HTML    | `pd.read_html()`    |
| Pickle  | `pd.read_pickle()`  |

### CSV

Como hemos visto en la clase anterior este es uno de los formatos más habituales en la industria. En general, todos los archivos `.csv` se leerán mediante el método `.read_csv()`. En las siguientes líneas exploraremos los argumentos más utilizados para la gestión de las distintas casuísticas.

Empecemos con el ejemplo más sencillo cargando una serie:

```python
serie_guardada = pd.read_csv("./data/serie.csv")
serie_guardada
```

Observamos que la serie no es capaz de distinguir el índice de otra columna y por tanto carga dos columnas (una con nombre y otra sin él) y genera un índice por defecto. Para evitar esta situación podemos usar el argumento `index_col` para indicar qué columna es el índice:

```python
serie_guardada = pd.read_csv("./data/serie.csv", index_col=0)
serie_guardada
```

```python
type(serie_guardada)
```

El método `pd.read_csv()` está pensado para DataFrames. En este caso si deseáramos cargarlo en una serie tendríamos que extraer la columna:

```python
serie_guardada = pd.read_csv("./data/serie.csv", index_col=0)["Serie guardar"]
serie_guardada
```

```python
type(serie_guardada)
```

Para leer un DataFrame podemos utilizar el mismo método de manera directa:

```python
df_guardado = pd.read_csv("./data/df_basic.csv", index_col=0)
df_guardado
```

El método asume que el separador que se está utilizando es la coma. Si se ha utilizado otro separador podremos obtener resultados raros como este:

```python
df_guardado = pd.read_csv("./data/df_sin_ind_sep.csv")
df_guardado
```

El argumento `sep` nos permite indicar el separador para solventar este problema:

```python
df_guardado = pd.read_csv("./data/df_sin_ind_sep.csv", sep=";")
df_guardado
```

```python
df_guardado.dtypes
```

Observamos que en este caso el método asume que la columna edad es de tipo string. Esto se debe a que no es capaz de reconocer la coma como separador decimal. El argumento `decimal` nos permite solucionar este problema:

```python
df_guardado = pd.read_csv("./data/df_sin_ind_sep.csv", sep=";", decimal=",")
df_guardado
```

```python
df_guardado.dtypes
```

Como hemos visto anteriormente en ocasiones la información puede almacenarse sin los nombres de columnas. Pandas __siempre__ tiene nombres en las columnas por lo que tenderá a inferirlos:

```python
df_guardado = pd.read_csv("./data/df_sin_col.csv", sep=";", decimal=",")
df_guardado
```

Este comportamiento puede generar errores y problemas por lo que en caso de no haber encabezados debemos indicarlo al momento de la lectura utilizando el atributo `header`:

```python
df_guardado = pd.read_csv("./data/df_sin_col.csv", sep=";", decimal=",", header=None)
df_guardado
```

Además el método nos permite colocar nuestros propios nombres sobre la marcha durante la lectura:

```python
df_guardado = pd.read_csv("./data/df_sin_col.csv", sep=";", decimal=",", 
                          names=["Nombre", "Coste", "Origen"])
df_guardado
```

Al igual que podíamos escribir solo algunas columnas dentro de un archivo, si sabemos que solo vamos a querer trabajar con algunas y queremos optimizar los recursos podemos utilizar el argumento `usecols` para utilizar esas columnas:

```python
df_guardado = pd.read_csv("./data/df_sin_ind_sep.csv", sep=";", decimal=",",
                         usecols=["nombre","ciudad"])
df_guardado
```

Por último, uno de los grandes problemas al cargar y leer información de archivos es la gestión de los tipos de datos. Este método cuenta con el argumento `dtype` que permite indicarle para cada columna en qué tipo queremos que se cargue:

```python
df_guardado = pd.read_csv("./data/df_sin_ind_sep.csv", sep=";", decimal=",",
                         dtype={
                             "edad" : np.float32
                         })
df_guardado
```

```python
df_guardado.dtypes
```
