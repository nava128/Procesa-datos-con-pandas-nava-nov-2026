# Sección 9. Diferencias entre NumPy y Pandas

Esta clase resume las diferencias más importantes entre **NumPy** y **Pandas**.

El objetivo de esta clase es consolidar el criterio para saber **cuándo usar cada una**, **qué problema resuelve mejor cada librería** y **cómo se complementan** dentro de un flujo de análisis de datos.


```python
import numpy as np
import pandas as pd

print("NumPy:", np.__version__)
print("Pandas:", pd.__version__)
```

## Objetivos de la clase

En esta clase vamos a:

- Comparar el propósito principal de NumPy y Pandas.
- Entender la diferencia entre los objetos `ndarray`, `Series` y `DataFrame`.
- Revisar cómo cambia el trabajo con índices, tipos de datos y valores nulos.
- Identificar en qué escenarios NumPy suele ser preferible y en cuáles Pandas aporta más valor.
- Cerrar el curso con una regla práctica de decisión.


La diferencia más importante puede resumirse así:

- **NumPy** está pensado para trabajar con **arrays homogéneos** y para realizar **cálculo numérico eficiente**.
- **Pandas** está pensado para trabajar con **datos tabulares etiquetados**, con índices, nombres de columnas y operaciones de análisis orientadas a datos reales.

Dicho de otra manera:

- Cuando el problema es principalmente **matemático o matricial**, NumPy suele ser la base natural.
- Cuando el problema es principalmente **analítico o tabular**, Pandas suele ofrecer una capa de trabajo más cómoda y expresiva.


### Estructuras de datos: `ndarray` frente a `Series` y `DataFrame`

NumPy y Pandas no compiten exactamente al mismo nivel de abstracción:

| Librería | Estructura principal | Idea clave |
|---|---|---|
| NumPy | `ndarray` | Array N-dimensional homogéneo |
| Pandas | `Series` | Vector unidimensional con índice |
| Pandas | `DataFrame` | Tabla bidimensional con filas y columnas etiquetadas |

En términos prácticos:

- Un `ndarray` prioriza el **dato y su forma**.
- Una `Series` añade al dato una **etiqueta de índice**.
- Un `DataFrame` organiza varias columnas bajo una misma estructura tabular.


```python
array_numpy = np.array([10, 20, 30, 40])
serie_pandas = pd.Series([10, 20, 30, 40], index=["Q1", "Q2", "Q3", "Q4"])
df_pandas = pd.DataFrame({
    "ventas": [10, 20, 30, 40],
    "costes": [7, 12, 18, 25]
}, index=["Q1", "Q2", "Q3", "Q4"])

print("Array de NumPy:")
print(array_numpy)
print("\nSeries de Pandas:")
print(serie_pandas)
print("\nDataFrame de Pandas:")
df_pandas
```

En los tres casos hay datos numéricos, pero el nivel de información contextual es muy distinto:

- En NumPy vemos valores y posición.
- En una `Series` vemos valores y etiquetas.
- En un `DataFrame` vemos una estructura tabular completa.

Ese contexto adicional es precisamente una de las razones por las que Pandas resulta tan útil en análisis de datos.


### Indexación: posición frente a etiquetas

En NumPy el acceso suele ser principalmente por **posición**.

En Pandas podemos trabajar tanto por **posición** como por **etiquetas**, lo que hace el código más legible cuando los datos representan entidades reales: fechas, nombres de variables, categorías, identificadores, etc.


```python
array_numpy = np.array([100, 200, 300, 400])
serie_pandas = pd.Series(
    [100, 200, 300, 400],
    index=["enero", "febrero", "marzo", "abril"]
)

print("NumPy por posición:", array_numpy[2])
print("Pandas por etiqueta:", serie_pandas["marzo"])
print("Pandas por posición con iloc:", serie_pandas.iloc[2])
```


- **NumPy** es excelente cuando la posición tiene significado suficiente.
- **Pandas** es más expresivo cuando necesitamos que el código refleje la semántica del dato.

### Homogeneidad de tipos frente a flexibilidad tabular

En NumPy, un `ndarray` contiene normalmente datos de un **único tipo** o, al menos, de un tipo común compatible.

En Pandas, cada columna de un `DataFrame` puede tener un tipo distinto. Esto es clave para representar tablas reales con números, texto, fechas o booleanos en la misma estructura.


```python
array_homogeneo = np.array([1, 2, 3, 4])
array_convertido = np.array([1, 2, 3.5, 4])  # promoción a float

df = pd.DataFrame({
    "producto": ["A", "B", "C"],
    "precio": [10.5, 20.0, 15.75],
    "stock": [100, 50, 80],
    "activo": [True, True, False]
})

print("dtype del array homogéneo:", array_homogeneo.dtype)
print("dtype del array convertido:", array_convertido.dtype)
print("\nTipos por columna en Pandas:")
print(df.dtypes)
```

- **NumPy** gana en simplicidad interna y eficiencia cuando todos los elementos responden al mismo esquema.
- **Pandas** gana en realismo cuando una tabla mezcla diferentes naturalezas de dato.

Por eso NumPy es la base natural del cálculo científico, mientras que Pandas encaja mejor en tareas de análisis, reporting y preparación de datos.


### Alineación automática de datos

Una diferencia muy característica de Pandas es la **alineación por índice**.

Cuando operamos entre objetos de Pandas, los resultados se alinean automáticamente por sus etiquetas. En NumPy, en cambio, las operaciones se basan en la forma y en la posición de los elementos.


```python
ventas_2024 = pd.Series([100, 120, 140], index=["enero", "febrero", "marzo"])
ventas_2025 = pd.Series([110, 130, 150], index=["febrero", "marzo", "abril"])

print("Suma alineada en Pandas:")
print(ventas_2024 + ventas_2025)

print("\nSuma posicional en NumPy:")
a = np.array([100, 120, 140])
b = np.array([110, 130, 150])
print(a + b)
```

La alineación automática es una ventaja enorme cuando trabajamos con datos etiquetados, pero también exige entender bien qué está ocurriendo:

- En **Pandas**, la semántica del índice importa.
- En **NumPy**, importa la forma del array.

Esto hace que Pandas sea más seguro y expresivo para análisis tabular, mientras que NumPy suele ser más directo en operaciones numéricas puras.


### Valores nulos y datos incompletos

Los datos reales suelen venir con ausencias, vacíos o registros incompletos.

Pandas está claramente más orientado a este problema. Ofrece una batería de operaciones para detectar, propagar, rellenar o excluir nulos. NumPy puede manejar ciertos valores especiales, pero no tiene una capa tan natural para el tratamiento tabular de datos faltantes.


```python
serie_con_nulos = pd.Series([10, None, 30, np.nan, 50])

print("Serie con nulos:")
print(serie_con_nulos)

print("\nDetección de nulos:")
print(serie_con_nulos.isna())

print("\nMedia ignorando nulos:")
print(serie_con_nulos.mean())

print("\nRelleno de nulos con 0:")
print(serie_con_nulos.fillna(0))
```

Si el problema incluye limpieza de datos, faltantes, registros incompletos o integración de fuentes heterogéneas, **Pandas suele ser la elección natural**.

NumPy puede intervenir después, cuando el objetivo pasa a ser el cálculo sobre datos ya estructurados.


### Agregación y análisis: Pandas ofrece una capa más cercana a negocio

NumPy dispone de funciones muy potentes de agregación sobre arrays. Sin embargo, Pandas añade una sintaxis mucho más próxima al análisis tabular: selección por columnas, resúmenes por variable, agrupación por categorías y operaciones descriptivas muy habituales.


```python
datos = pd.DataFrame({
    "region": ["Norte", "Norte", "Sur", "Sur", "Centro", "Centro"],
    "ventas": [120, 150, 90, 110, 130, 160],
    "costes": [80, 95, 70, 75, 85, 100]
})

print("Media de ventas con NumPy:")
print(np.mean(datos["ventas"].to_numpy()))

print("\nMedia de ventas con Pandas:")
print(datos["ventas"].mean())

print("\nVentas medias por región con Pandas:")
print(datos.groupby("region")["ventas"].mean())
```

- **NumPy** agrega muy bien sobre arrays.
- **Pandas** agrega muy bien sobre estructuras tabulares con significado analítico.

Cuando aparece una pregunta del tipo "¿cuál es la media por categoría?" o "¿qué ocurre por región, cliente o producto?", Pandas suele reducir mucho la complejidad del código.


### Rendimiento: qué significa realmente que NumPy sea más rápido

Se suele decir que NumPy es más rápido que Pandas. Como regla general, esto es razonable, pero conviene matizarlo:

- NumPy suele ser más eficiente en **cálculo numérico puro**.
- Pandas añade capas de abstracción muy útiles: índices, etiquetas, validaciones, alineación y métodos orientados al análisis.
- Esa comodidad adicional puede tener un coste, pero muchas veces compensa por claridad y productividad.

La comparación correcta no es solo **velocidad bruta**, sino **velocidad total de resolver bien el problema**.


```python
matriz = np.arange(12).reshape(3, 4)
df = pd.DataFrame(matriz, columns=["A", "B", "C", "D"])

print("Suma total con NumPy:")
print(matriz.sum())

print("\nSuma por columnas con Pandas:")
print(df.sum())

print("\nConversión de Pandas a NumPy:")
print(df.to_numpy())
```

- Si todo el problema es numérico y estructurado, **NumPy suele ser más ligero**.
- Si el problema incluye contexto tabular, nombres de columnas, filtrado, nulos o agrupaciones, **Pandas suele ser más productivo**.

### Relación entre ambas librerías

No conviene pensar en NumPy y Pandas como dos alternativas totalmente excluyentes.

En realidad, lo habitual es que trabajen juntas:

1. **Pandas** organiza y transforma los datos.
2. **NumPy** aporta la base numérica y operaciones eficientes sobre arrays.
3. En muchos casos, un objeto de Pandas termina convirtiéndose a NumPy para alimentar modelos, funciones numéricas o bibliotecas externas.


```python
df_metricas = pd.DataFrame({
    "x": [1, 2, 3, 4],
    "y": [10, 20, 30, 40]
})

valores_numpy = df_metricas[["x", "y"]].to_numpy()

print("DataFrame original:")
print(df_metricas)

print("\nArray convertido desde Pandas:")
print(valores_numpy)
print(type(valores_numpy))
```

## ¿Cuándo usar cada una?

### Usaría principalmente **NumPy** cuando:

- trabajo con cálculo matricial o vectorial,
- necesito arrays homogéneos,
- estoy en un contexto científico o numérico puro,
- quiero alimentar algoritmos que operan sobre arrays,
- el foco principal está en rendimiento y estructura regular.

### Usaría principalmente **Pandas** cuando:

- trabajo con tablas de datos reales,
- necesito nombres de columnas e índices significativos,
- hay valores nulos o datos incompletos,
- debo filtrar, agrupar, agregar o combinar datos,
- el objetivo es análisis, exploración o preparación de datos.


La comparación puede cerrarse con esta tabla resumen:

| Criterio | NumPy | Pandas |
|---|---|---|
| Enfoque principal | Cálculo numérico | Análisis tabular |
| Estructura base | `ndarray` | `Series` y `DataFrame` |
| Tipos de datos | Homogéneos | Heterogéneos por columna |
| Indexación | Posicional | Posicional y por etiquetas |
| Nulos | Soporte más limitado | Soporte muy natural |
| Agrupación tabular | Más manual | Muy potente |
| Lectura de datos reales | Más limitada | Muy completa |
| Rendimiento numérico puro | Muy alto | Bueno, con más abstracción |

La idea más importante del curso es esta:

> **NumPy optimiza el cálculo sobre arrays. Pandas optimiza el trabajo con datos tabulares.**

No son librerías excluyentes, sino complementarias. Saber diferenciarlas bien es una de las habilidades básicas para trabajar con datos con criterio.
