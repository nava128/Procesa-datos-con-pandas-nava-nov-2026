# Sección 7. Agregación de datos

En esta sección, vamos a estudiar la agregación de datos que nos permite condensar la información más relevante de un DataFrame así como explorar de una manera más general las variables y sus distribuciones.

## 7.1. Introducción a la agregación de datos

__Def.__ La __agregación de datos__ consiste en resumir datos agrupando la información de distintas filas mediante funciones estadísticas.

Esto nos permite obtener unos valores resumidos que nos ayudan a comprender de manera más general la información.

La agregación responde a preguntas clave como los promedios de un valor repartido por una categoría o la suma de una determinada variable para cada valor de dicha categoría. Algunos ejemplos de este tipo de preguntas muy habituales en análisis empresarial y ciencia de datos son:

* ¿Cuál es el promedio de salario por ciudad?
* ¿Cuántos clientes hay por país?
* ¿Cuáles son las ventas totales por producto?

## Split-Apply-Combine

La agregación en Pandas se conforma sobre el paradigma __Split-Apply-Combine__ que exploraremos a continuación.

Partamos de un conjunto de datos muy sencillo:

| ciudad  | ventas |
| ------- | ------ |
| Madrid  | 100    |
| Madrid  | 200    |
| Sevilla | 150    |
| Sevilla | 120    |


### Split (dividir)

Dividimos el dataset tomando una (o varias) variables como eje. En este caso vamos a utilizar la variable `ciudad` que nos generará dos grupos:

| ciudad  | ventas |
| ------- | ------ |
| Madrid  | 100    |
| Madrid  | 200    |


| ciudad  | ventas |
| ------- | ------ |
| Sevilla | 150    |
| Sevilla | 120    |

### Apply (aplicar)

Aplicamos nuestra función de agregación a cada uno de los grupos. La función de agregación elegida dependerá de la información que estemos buscando. Por ejemplo, si queremos responder a la pregunta ¿cuál ha sido la mayor venta alcanzada en cada ciudad? utilizaríamos el máximo mientras que si quisiéramos responder a la pregunta ¿cuánto se ha vendido en total en cada ciudad? utilizaríamos la suma y obtendríamos:

| ciudad  | ventas |
| ------- | ------ |
| Madrid  | 300    |


| ciudad  | ventas |
| ------- | ------ |
| Sevilla | 270    |

### Combine (combinar)

Combinamos los resultados obtenidos en cada grupo en una única estructura final que __agrega__ toda la información:

| ciudad  | ventas |
| ------- | ------ |
| Madrid  | 300    |
| Sevilla | 270    |

Esta estructura respondería de manera directa a la pregunta ¿cuánto se ha vendido en total en cada ciudad?

### Implementación en Pandas

Una vez sentada la base teórica vamos a estudiar el método `.groupby()` de Pandas que automatiza este cálculo de manera sencilla. Empezamos creando nuestro DataFrame en Pandas:

```python
import pandas as pd

data = {
    "ciudad": ["Madrid", "Madrid", "Sevilla", "Sevilla"],
    "ventas": [100, 200, 150, 120]
}

df_ventas = pd.DataFrame(data)
df_ventas
```

El método `.groupby()` recibe una o varias columnas como argumento y genera la segregación en grupos. Estos grupos no son legibles si no agregamos una función de agregación. Por ejemplo, la suma:

```python
df_ventas.groupby("ciudad")
```

```python
df_agregado = df_ventas.groupby("ciudad").sum()
df_agregado
```

Hemos obtenido de manera rápida y simple nuestro DataFrame agregado con el total de ventas para cada categoría.

### Funciones de agregación más habituales

En este apartado veremos algunas de las funciones de agregación más utilizadas al trabajar con Pandas.

#### media

Respondería a la pregunta ¿cuáles son las ventas promedio por ciudad?

```python
df_ventas.groupby("ciudad").mean()
```

#### mediana

Respondería a la pregunta ¿cuál es la venta mediana por cada ciudad?

```python
df_ventas.groupby("ciudad").median()
```

#### mínimo

Respondería a la pregunta ¿cuál es la venta mínima en cada ciudad?

```python
df_ventas.groupby("ciudad").min()
```

#### máximo

Respondería a la pregunta ¿cuál es la mayor venta producida en cada ciudad?

```python
df_ventas.groupby("ciudad").max()
```

#### conteo

Respondería a la pregunta ¿cuántos registros tenemos por cada ciudad?

```python
df_ventas.groupby("ciudad").count()
```

#### desviación estándar y varianza

Respondería a la pregunta ¿cuál es la variabilida de la distribución asociada a cada ciudad?

```python
df_ventas.groupby("ciudad").std()
```

```python
df_ventas.groupby("ciudad").var()
```

#### cuantiles

Respondería a la pregunta ¿cuál es el cuantil n de la distribución asociada a cada ciudad?

```python
df_ventas.groupby("ciudad").quantile(0.1)
```

```python
df_ventas.groupby("ciudad").quantile(0.9)
```

__Nota.__ Emplea un método numérico de aproximación lo que hace que arroje resultados extraños para estos ejemplos en los que solo trabajamos con dos variables.

#### valores únicos

Respondería a la pregunta ¿cuántos valores distintos de ventas tengo por cada ciudad?

```python
df_ventas.groupby("ciudad").nunique()
```

### 7.2. Agregaciones múltiples en Pandas

Hasta ahora todas las agregaciones que hemos visto nos permiten agregar operaciones por una única columna. En esta clase veremos como agrupar datos por varias columnas simultáneamente y las jerarquías que esto genera. 

La clave de esta idea es que nos permitirá analizar datos desde varias dimensiones al mismo tiempo.

Por ejemplo, partiendo de la siguiente tabla:

| ciudad   | producto | categoria  | ventas | precio | unidades |
| -------- | -------- | ---------- | ------ | ------ | -------- |
| Madrid   | A        | Tecnologia | 120    | 10     | 12       |
| Madrid   | B        | Tecnologia | 200    | 20     | 10       |
| Madrid   | A        | Hogar      | 150    | 15     | 10       |
| Sevilla  | A        | Hogar      | 180    | 12     | 15       |
| Sevilla  | B        | Tecnologia | 130    | 22     | 6        |
| Valencia | A        | Hogar      | 160    | 14     | 11       |
| Valencia | B        | Tecnologia | 170    | 21     | 8        |


Podemos agregar, por ejemplo, por ciudad y categoría obteniendo los grupos: Madrid-Tecnología, Madrid-Hogar, Sevilla-Hogar, Sevilla-Tecnología, Valencia-Hogar y Valencia-Tecnología.

Veamos un ejemplo de cómo podemos hacer esto en Pandas:

```python
import pandas as pd

data = {
    "ciudad": ["Madrid", "Madrid", "Madrid", "Sevilla", "Sevilla", "Valencia", "Valencia"],
    "producto": ["A", "B", "A", "A", "B", "A", "B"],
    "categoria": ["Tecnologia", "Tecnologia", "Hogar", "Hogar", "Tecnologia", "Hogar", "Tecnologia"],
    "ventas": [120, 200, 150, 180, 130, 160, 170],
    "precio": [10, 20, 15, 12, 22, 14, 21],
    "unidades": [12, 10, 10, 15, 6, 11, 8]
}

df_ventas = pd.DataFrame(data)

df_ventas
```

```python
df_agreg_ciudad_cat = df_ventas.groupby(["ciudad", "categoria"])["ventas"].sum()
df_agreg_ciudad_cat
```

El resultado anterior contiene un `MultiIndex`, es decir, un índice con varios niveles. En este caso tenemos en el nivel 1 la ciudad y en el nivel 2 el producto. Esto nos permite organizar datos multidimensionales:

```python
df_agreg_ciudad_cat.index
```

Mediante el método `.loc` podemos acceder a un grupo específico, dentro del primer nivel, por ejemplo:

```python
df_agreg_ciudad_cat.loc["Madrid"]
```

Pero con el mismo método también podemos acceder al nivel más bajo:

```python
df_agreg_ciudad_cat.loc[("Madrid", "Hogar")]
```

Estos métodos se amplían si añadimos nuevas dimensiones de agregación:

```python
df_agreg_triple = df_ventas.groupby(["ciudad", "producto", "categoria"])["ventas"].sum()
df_agreg_triple
```

```python
df_agreg_triple.loc["Madrid"]
```

```python
df_agreg_triple.loc[("Madrid", "A")]
```

```python
df_agreg_triple.loc[("Madrid", "A", "Hogar")]
```

Además, es interesante ver cómo no podemos utilizar también un acceso directo saltando la jerarquía intermedia de producto:

```python
df_agreg_triple.loc[("Madrid", "Hogar")]
```

Por último es interesante ver que utilizando el método `.reset_index()` nos permite aplanar nuestro multiíndice en una tabla:

```python
df_agreg_triple
```

```python
df_agreg_triple.reset_index()
```

Esto puede resultar útil en ocasiones porque mejora la lejibilidad de la información.

## 7.3. Uso de `.agg()` y agregación con funciones personalizadas

### Introducción al método `.agg()`

En esta clase vamos a explorar la función `.agg()` que nos permite realizar y combinar agrupaciones de manera más versátil.

El método `.agg()` permite aplicar una o varias funciones de agregación. 

Es decir con este método podemos aplicar múltiples agregaciones a una misma columna. Veamos un ejemplo:




```python
import pandas as pd

data = {
    "ciudad": ["Madrid", "Madrid", "Madrid", "Sevilla", "Sevilla", "Valencia", "Valencia"],
    "producto": ["A", "B", "A", "A", "B", "A", "B"],
    "ventas": [120, 200, 150, 180, 130, 160, 170],
    "precio": [10, 20, 15, 12, 22, 14, 21],
    "unidades": [12, 10, 10, 15, 6, 11, 8]
}

df = pd.DataFrame(data)
df
```

```python
df.groupby("ciudad")["ventas"].agg(["sum", "mean", "max"])
```

El resultado es un DataFrame que contiene las tres agregaciones (suma, media y máximo) de la variable ventas por ciudad.

Además, el método `.agg()` nos permite especificar agregaciones diferentes para cada columna de manera que generamos un DataFrame con exactamente las agregaciones que necesitamos. Para llevar a cabo esta operación utilizaremos como argumento un diccionario que tiene por clave la variable sobre la que queremos realizar la agregación y como valor la función (o lista de funciones) que deseamos aplicar:

```python
dic_agregaciones = {
    "ventas": ["sum", "mean"],
    "precio": "mean",
    "unidades": "sum"
}

df.groupby("ciudad").agg(dic_agregaciones)
```

La parte negativa de realizar las agregaciones de esta manera es que los nombres de las columnas pueden resultar en ocasiones salgo complicados de leer e intuir. Podemos indicar dentro del propio método cómo queremos denominar a la variable resultante:

```python
df.groupby("ciudad").agg(
    ventas_totales=("ventas", "sum"),
    ventas_promedio=("ventas", "mean"),
    precio_promedio=("precio", "mean"),
    unidades_totales=("unidades", "sum")
)
```

De esta manera obtenemos unos resultados mucho más claros especialmente si pensamos utilizarlos a posteriori en presentaciones o documentos que serán consumidos por usuarios no técnicos.

Además, el método `.agg()` nos permite trabajar con agrupaciones sobre múltiples categorías. Veamos un ejemplo en el que utilizamos dos columnas como categorías y varias funciones de agregación:

```python
df_agg_ciudad_prod = df.groupby(["ciudad", "producto"]).agg(
    ventas_totales=("ventas", "sum"),
    ventas_promedio=("ventas", "mean"),
    precio_max=("precio", "max")
)
df_agg_ciudad_prod
```

Recordemos además, que utilizando el método `.reset_index()` podemos transformar este resultado en una tabla plana:

```python
df_agg_ciudad_prod.reset_index()
```

### Agregaciones personalizadas

__Def.__ Una __agregación personalizada__ es una función definida por el usuario que se aplica a un grupo de datos para producir un valor resumido.

Las funciones de agregación estandar como la media o la suma no siempre son suficientes y en ocasiones querremos realizar cálculos más complejos adaptados a las necesidades del análisis que estamos llevando a cabo.

Para ello lo primero que debemos hacer es definir nuestra función de agregación, por ejemplo, supongamos que queremos calcular el rango de una distribución, es decir, la diferencia entre el máximo y el mínimo de dicha distribución. Puede ser interesante ver en qué ciudades hay mayor variabilidad de ventas y cómo de marcada está.

Empezamos definiendo nuestra función que recibirá una serie como argumento y devolverá la diferencia entre su valor máximo y su valor mínimo:



```python
def calcula_rango(series):
    return series.max() - series.min()
```

Para utilizarla como función de agregación la usaremos como argumento del método `.agg()`:

```python
df.groupby("ciudad")["ventas"].agg(calcula_rango)
```

Esta función al igual que las funciones de agregaciones tradicionales es susceptible de ser utilizada en varias columnas y de generar resultados renombrados:

```python
df.groupby(["ciudad"]).agg(
    rango_ventas=("ventas", calcula_rango),
    rango_precio=("precio", calcula_rango)
)
```

Análogamente podemos utilizar funciones de agregación personalizadas combinadas con las funciones de agregación predefinidas:

```python
df.groupby("ciudad")["ventas"].agg(
    ["sum", "mean", calcula_rango]
)
```

En caso de que no tengamos pensado utilizar más la función de agregación podemos definirla directamente como una función anónima sobre todo si la lógica es tan sencilla como en este caso. 

```python
df.groupby("ciudad")["ventas"].agg(lambda x: x.max() - x.min())
```

Por último es interesante ver que una función de agregación no tiene por qué devolver un único valor si no que puede ser definida para devolver varios valores, por ejemplo:

```python
def calcula_cuartiles(serie):
    return pd.Series({
        "primer_cuartil" : serie.quantile(0.25),
        "segundo_cuartil" : serie.median(),
        "tercer_cuartil" : serie.quantile(0.75),
        "cuarto_cuartil" : serie.max()
    })
```

Para usar esta función necesitamos un DataFrame con más valores:

```python
import numpy as np

ciudades = ["Madrid", "Oviedo", "Sevilla"]

data = {
    "ciudad": np.repeat(ciudades, 15),
    "mm_lluvia": np.concatenate([
        np.random.uniform(0, 20, 15),   # Madrid
        np.random.uniform(10, 80, 15),  # Oviedo (más lluvia)
        np.random.uniform(0, 10, 15)    # Sevilla (menos lluvia)
    ])
}

df_lluvia = pd.DataFrame(data)
df_lluvia.head()
```

Calculamos la agregación con los cuartiles:

```python
df_lluvia.groupby("ciudad")["mm_lluvia"].apply(calcula_cuartiles)
```

## 7.4. Uso de transformaciones

En esta clase vamos a explorar el método `.transform()` y sus diferencias con `.agg()`.

__Def.__ Una __transformación__ es una operación que se calcula por grupo y devuelve un resultado del mismo tamaño que el DataFrame original.

Es decir, `.agg()` nos permitía reducir los datos mientras que `.transform()` mantiene el número de registros.

Las transformaciones son especialmente útiles para la creación de columnas derivadas, por ejemplo, calculamos el salario promedio por departamento y lo añadimos en una columna para cada empleado para comparar su salario con el salario medio o calcular la diferencia y ver si está "infrapagado" o "sobrepagado".

Veamos un ejemplo práctico retomando nuestro dataset de ventas:

```python
import pandas as pd

data = {
    "ciudad": ["Madrid", "Madrid", "Madrid", "Sevilla", "Sevilla", "Valencia", "Valencia"],
    "producto": ["A", "B", "A", "A", "B", "A", "B"],
    "ventas": [120, 200, 150, 180, 130, 160, 170],
    "precio": [10, 20, 15, 12, 22, 14, 21]
}

df_ventas = pd.DataFrame(data)

df_ventas
```

Para calcular una transformación, por ejemplo, la media de ventas por ciudad usaremos el método `.transform()` análogamente a como hacíamos con `.agg()` indicando la operación que deseamos aplicar:

```python
df_ventas.groupby("ciudad")["ventas"].transform("mean")
```

El resultado tiene el mismo número de filas que el DataFrame original por lo que podemos almacenar este valor en una nueva columan de nuestro DataFrame original:

```python
df_ventas["media_ventas_ciudad"] = df_ventas.groupby("ciudad")["ventas"].transform("mean")
df_ventas
```

Observamos que tenemos la media calculada por ciudad y que esto nos permite comparar qué registros incluyen ventas por encima y por debajo de la media e incuso cuantificar esa diferencia:

```python
df_ventas["diferencia_media"] = df_ventas["ventas"] - df_ventas.groupby("ciudad")["ventas"].transform("mean")
df_ventas
```

Al igual que `.agg()`, `.transform()` nos permite realizar transformaciones a partir de funciones personalizadas creadas por nosotros. Volviendo al ejemplo del intervalo:

```python
df_ventas["intervalo_ventas_ciudad"] = df_ventas.groupby("ciudad")["ventas"].transform(lambda elemento: elemento.max() - elemento.min())
df_ventas
```

Cerraremos esta clase viendo de manera clara la diferencia entre una función de agregación y una de transformación:

| Característica           | `agg()`   | `transform()`   |
| ------------------------ | --------- | --------------- |
| Reduce datos             | Sí        | No              |
| Mantiene tamaño original | No        | Sí              |
| Uso típico               | resúmenes | nuevas columnas |


```python
df_ventas.groupby("ciudad")["ventas"].agg("sum")
```

```python
df_ventas.groupby("ciudad")["ventas"].transform("sum")
```

Observamos que los valores son los mismos pero mientras que en un caso se "agregan" a un único registro en otro se mantiene la correspondencia índice-valor.
