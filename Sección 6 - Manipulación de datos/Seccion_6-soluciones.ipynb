# Sección 6. Manipulación de datos

En esta sección vamos a ver distintas técnicas para manipular datos, como la generación de columnas a partir del valor de columnas ya existentes o la aplicación de funciones a distintas columnas y filas de nuestro DataFrame.

## 6.1. Creación y eliminación de columnas y filas. Modificación de valores

Los DataFrames son estructuras mutables lo que significa que podemos hacer modificaciones como: 

* Añadir o eliminar columnas.
* Añadir o eliminar filas.
* Modificar valores ya existentes en el DataFrame.

En esta clase veremos técnicas para llevar a cabo estas modificaciones de manera controlada sin poner en riesgo la integridad y coherencia de nuestros datos.

Es importante enteder que la modificación de datos es una operación fundamental a nivel funcional ya que nos permite corregir valores erróneos, limpiar registros o crear variables derivadas de las ya existentes.

### Creación de columnas

En clases previas ya hemos visto dos maneras de crear columnas.

#### Creación de columnas con datos constantes

Simplemente instanciamos la nueva columna y le asociamos el valor constante deseado. Pandas replicará dicho valor a lo largo de toda la columna:

```python
import pandas as pd

data = {
    "nombre": ["Ana", "Diana", "Carlos", "Cris"],
    "edad": [25, 30, 22, 20],
    "ciudad": ["Oviedo", "Sevilla", "Madrid", "Bilbao"],
    "salario": [3500, 2000, 1200, 1250]
}

df_columnas = pd.DataFrame(data)
df_columnas
```

```python
df_columnas["país"] = "España"
df_columnas
```

Además, vimos que podemos insertar la columna en una posición específica usando el método `.insert()`:

```python
df_columnas.insert(4, "continente", "europa")
df_columnas
```

#### Creación de columnas a partir de datos existentes

Una vez contamos con datos en un DataFrame en muchas ocasiones vamos a querer realizar transformaciones a partir de una o más variables existentes. Pandas y la vectorización de operaciones heredadas de NumPy nos permiten hacer esto de manera sencilla:

```python
df_columnas["edad_meses"] = df_columnas["edad"] * 12
df_columnas
```

Esta operación es también compatible con el uso de `.insert()` permitiéndonos crear dicha columna en la posición que más nos convenga:

```python
df_columnas.insert(2, "edad_dias", df_columnas.edad*365)
df_columnas
```

Análogamente a la creación de máscaras podemos crear columnas con booleanos que verifiquen una determinada condición:

```python
df_columnas["acaba_en_a"] = df_columnas.nombre.str.endswith("a")
df_columnas
```

#### Modificación de valores en una columna

Una vez creada una columna, por ser una estructura mutable podemos realizar modificaciones sobre ella, por ejemplo:

```python
df_columnas["continente"] = df_columnas["continente"].str.capitalize()
df_columnas
```

En este caso hemos realizado una modificación completa de la columna pero en ocasiones querremos realizar modificaciones solo en algunos casos determinados. Pandas implementa la posbilidad de realizar modificaciones condicionales mediante el método `.loc`. Por ejemplo, si quisiéramos asumir una edad mínima para nuestros datos:

```python
df_columnas.loc[df_columnas["edad"] < 25, "edad"] = 24
df_columnas
```

En este caso hemos modificado los valores de Carlos y Cris que estaban por debajo de 25 y hemos dejado los demás iguales.

El uso de `.loc` es una buena práctica que garantiza la integridad de los datos. La modificación mediante el acceso directo a índices es posible en ocasiones aunque da error en muchos casos y suele generar warnings:

```python
df_columnas["edad"][0] = 26
df_columnas
```

### Creación de filas

Para la creación de una nueva fila sobre un DataFrame podemos emplear el método `.loc`. Podemos o bien insertar los valores a mano:

```python
df_columnas.loc[df_columnas.shape[0]] = ["Pedro", 28, 28*365, "Bilbao", 3000, "Europa", "España", 28*12, False]
df_columnas
```

Este método es útil para DataFrames pequeños pero si nuestro DataFrame tiene bastantes columnas puede ser más sencillo utilizar un diccionario:

```python
nueva_fila = {
    "nombre" : "Arturo",
    "edad" : 31,
    "ciudad" : "Oviedo",
    "continente" : "Europa"
}
```

```python
df_columnas.loc[len(df_columnas)] = nueva_fila
df_columnas
```

__Nota.__ Observamos que este método además nos permite indicar solo los valores que nos interesan o que conocemos y rellena de manera automática el resto de valores con valores nulos.

### Eliminación de columnas y filas

El método `drop()` nos resulta de especial utilidad cuando queremos eliminar determinadas filas o columnas. El argumento `axis` que toma por defecto el valor 0 (filas), nos permite decidir qué estructura eliminamos.

Podemos eliminar según deseemos una o varias columnas:

```python
df_columnas.drop("edad_dias", axis=1)
```

```python
df_columnas.drop(["edad_dias", "edad_meses"], axis=1)
```

__Nota.__ Recordemos que la operación no se realiza en el propio objeto por lo que si no almacenamos el resultado sobre el objeto o un nuevo objeto la eliminación no se persiste:

```python
df_columnas
```

El proceso es exactamente análogo para filas. Podemos no explicitar el argumento `axis` pues como hemos visto toma el valor 0 por defecto:

```python
df_columnas.drop(0)
```

```python
df_columnas.drop([0,2])
```

__Nota.__ De nuevo la operación no se persiste:

```python
df_columnas
```

Por último es interesante ver que podemos eliminar filas por condiciones. Esto sería el equivalente al filtrado que hemos visto en la sección anterior:

```python
df_columnas_mayores = df_columnas[df_columnas["edad"] >= 25]
df_columnas_mayores
```

## 6.2. Funciones con apply

En esta clase vamos a estudiar el método `apply()` y su funcionamiento sobre Series y DataFrames.

El método `apply()` permite __aplicar una función a cada elemento o estructura de datos__ dentro de una Series o un DataFrame.

Es posible aplicar este método tanto elemento a elemento como a cada fila o columna en el caso de tratarse de un DataFrame.

`apply()` se utiliza cuando queremos aplicar una transformación personalizda que no puede realizarse de manera sencilla con alguna operación vectorizada como las estudiadas hasta ahora, por ejemplo, aplicando reglas personalizadas o funciones matemáticas complejas.

### `apply()` en series

Veamos un ejemplo muy sencillo para empezar:

```python
import pandas as pd

serie_ejemplo = pd.Series([1, 2, 3, 4])
serie_ejemplo
```

Utilizamos una función lambda en este caso para definir una operación:

```python
serie_duplicada = serie_ejemplo.apply(lambda elemento: elemento * 2)
serie_duplicada
```

En esta función indicamos que queremos sustituir cada elemento de la serie por dicho elemento multiplicado por 2. Esto sería equivalente a realizar:

```python
serie_ejemplo * 2
```

En este caso `apply()` no está aportando ninguna funcionalidad nueva pero que pasa si tenemos un polinomio que nos ayuda a estimar los ingresos de una persona según su edad:

```python
def estima_ingresos(edad):
    return 0.5*(edad**2) - 5*edad + 1000
```

No existe una función vectorizada para calcular exactamente este logartimo por lo que en este caso el método `apply()` resulta de gran utilidad:

```python
serie_edades = pd.Series([18, 25, 40, 56, 78])
serie_edades
```

```python
serie_ingresos = serie_edades.apply(lambda years: estima_ingresos(years))
serie_ingresos
```

### `apply()` en DataFrames

Al pasar de un elemento unidimensional a un elemento bidimensional, en los DataFrames deberemos utilizar el argumento `axis` con el que ya estamos bastante familiariados para decidir si la función se aplica por filas o por columnas.

Veamos un ejemplo práctico:

```python
data = {
    "edad": [25, 30, 22],
    "salario_mes": [2000, 3000, 2500]
}

df_salario = pd.DataFrame(data)
df_salario
```

```python
df_salario["salario_anual"] = df_salario.apply(
    lambda fila: fila["salario_mes"] * 12,axis=1)
df_salario
```

En este caso cada elemento que entra en la función es una fila. Podemos acceder a un elemento concreto de la fila como si fuera un diccionario utilizando el nombre de la columna como clave. Una vez hecho eso, aplicamos la operación de manera que generamos una serie de tres elementos. Dicha serie se asocia a la columna `salario_anual`.

Es interesante ver que podemos basarnos en más de una columna para nuestro cálculo:

```python
df_salario["numero_pagas"] = df_salario.apply(
    lambda fila: fila["salario_anual"] // fila["salario_mes"],axis=1)
df_salario
```

Análogamente, podemos hacer cálculos a nivel fila, por ejemplo, obtener los totales:

```python
df_salario.apply( lambda columna: columna.sum())
```

__Nota.__ En este caso omitimos el argumento `axis` ya que toma el valor 0 por defecto que indica que la operación debe hacerse por filas.

Uno de los usos más populares de `apply()` es su uso junto a funciones de "bucketización" que nos permiten agrupar la información de variables numéricas en variables categóricas:

```python
import numpy as np
df_ejemplo = pd.DataFrame(data=np.random.randint(1,100, (20,4)), 
                          columns=["A", "B", "C", "D"]
                         )
df_ejemplo
```

```python
def categoriza_valor(numero):
    if numero > 70:
        return "alto"
    elif numero > 30:
        return "medio"
    else:
        return "bajo"
```

```python
df_ejemplo["nivel_A"] = df_ejemplo.apply(
    lambda fila: categoriza_valor(fila["A"]), axis=1)
df_ejemplo
```

### Rendimiento en `apply()`

`apply()` ejecuta las funciones elemento a elemento en Python lo que implica una mayor sobrecarga de memoria y por tanto un menor rendimiento.

El uso de operaciones vectorizadas por el contrario, como hemos visto previamente, utiliza código optimizado en C para operar sobre arrays completos.

__Siempre que podamos debemos usar una operación vectorizada en lugar de__ `apply()`.

`apply()` debe utilizarse únicamente cuando:

* la operación no puede vectorizarse fácilmente.
* necesitamos una lógica condicional muy compleja.
* aplicamos funciones personalizadas.
* transformaremos datos fila a fila.

## 6.3. `map()`  en Pandas

En esta clase vamos a estudiar `map()`un método que nos permite aplicar funciones elemento a elemento sobre el total de la estructura de datos con la que trabajamos.

El método `map()` se utilizaba solo en Series pero en versiones más recientes de Pandas se extiende también a los DataFrames y nos permite transformar cada elemento de la serie. Podemos usarlo de manera análoga a `apply()`:

```python
import pandas as pd 

ciudades = pd.Series(["Oviedo", "Sevilla", "Bilbao"])
ciudades
```

```python
ciudades_mayus = ciudades.map(str.upper)
ciudades_mayus
```

En este caso como ya sabemos que la función se va a aplicar a todos los elementos la podemos indicar directamente. Además, `map()` nos permite usar diccionarios para sustituciones.

Por ejemplo:

```python
dic_autonomias = { "Oviedo" : "Asturies",
                   "Sevilla" : "Andalucía",
                   "Bilbao" : "País Vasco"
                 }
dic_autonomias
```

```python
ciudades.map(dic_autonomias)
```

Además, en caso de no enocntrar un valor genera un valor nulo:

```python
ciudades[3] = "Barcelona"
ciudades
```

```python
ciudades.map(dic_autonomias)
```

El funcionamiento es exactamente análogo para `DataFrames`:

```python
data = {
    "Dias_Prestado": [5, 12, 7, 16, 20, 10, 4, 6, 15, 9, 2, 18],
    "Días_Lectura": [4, 5, 2, 4, 15, 3, 4, 2, 5, 4, 2, 11]
}

df_libros = pd.DataFrame(data)
df_libros
```

Si quisiéramos trabajar con la información en semanas en lugar de días:

```python
df_libros_semanas = df_libros.map(lambda dias: dias/7)
df_libros_semanas
```

### Impacto en rendimiento

De nuevo `.map()` no aprovecha la vectorización de la que dispone Pnadas por lo que resulta menos eficiente. Sin embargo, resulta útil cuando:

* Deseamos recodificar valores
* Aplicar funciones definidas por nosotros mismos a **todos** los elementos.
* Convertir formatos
* Aplicación de funciones no vectorizables.

Como regla general siempre que podamos usar operaciones vectorizadas debemos evitar utilizar `apply()` y `map`.

## 6.4. Tratamiento de valores nulos

En esta clase vamos a continuar el camino que iniciamos con los ejemplos de Series y ver cómo se realiza el tratamiento de nulos dentro de un DataFrame.

Recordemos que un valor nulo representa la ausencia de información mediante un valor indeterminado (NaN).

Estos valores representan datos no disponibles, campos opcionales, errores en la captura de datos, fallos en sensores...

Es importante entender que un valor nulo en ocasiones puede indicar en sí información, por ejemplo, en un censo electoral la variable `fecha_fallecimiento` solo contendrá valores para las personas fallecidas y presentará nulos en los casos de personas vivas.

### Detección de valores nulos

Empecemos viendo cómo podemos utilizar Pandas para detectar valores nulos.

Partimos del siguiente DataFrame de ejemplo:

```python
import pandas as pd
import numpy as np

data = {
    "ciudad": [
        "Madrid","Barcelona","Sevilla","Valencia","Bilbao",
        "Paris","Berlin","Roma","Lisboa","Londres",
        "Buenos Aires","Santiago","Bogotá","Ciudad de México","Montevideo",
        "Quito","La Paz","Caracas","Lima","Asunción",
        "Tokio","Seúl","Bangkok","Singapur","Jakarta"
    ],
    
    "pais": [
        "España","España","España","España","España",
        "Francia","Alemania","Italia","Portugal","Reino Unido",
        "Argentina","Chile","Colombia","México","Uruguay",
        "Ecuador","Bolivia","Venezuela","Perú","Paraguay",
        "Japón","Corea del Sur","Tailandia","Singapur","Indonesia"
    ],
    
    "continente": [
        "Europa","Europa","Europa","Europa","Europa",
        "Europa","Europa","Europa","Europa","Europa",
        "América","América","América","América","América",
        "América","América","América","América","América",
        "Asia","Asia","Asia","Asia","Asia"
    ],
    
    "temperatura_media": [
        22,24,np.nan,26,18,
        20,np.nan,27,25,16,
        23,19,18,21,np.nan,
        17,12,np.nan,20,24,
        18,17,29,28,np.nan
    ],
    
    "precipitacion_mm": [
        420,None,520,450,1200,
        np.nan,570,800,750,None,
        1100,300,850,720,950,
        None,650,900,10,np.nan,
        1500,1400,None,2300,2000
    ],
    
    "humedad_media": [
        60,65,55,None,70,
        75,np.nan,68,72,80,
        77,65,78,None,74,
        82,np.nan,79,83,76,
        68,70,85,None,88
    ],
    
    "estacion_meteorologica": [
        "Centro","Puerto",None,"Aeropuerto","Costa",
        "Norte","Este",None,"Oeste","Centro",
        "Sur","Valle","Altura",None,"Puerto",
        "Andes","Altiplano",None,"Costa","Interior",
        "Metro","Capital","Tropical","Ciudad",None
    ]
}

df_nulos = pd.DataFrame(data)
df_nulos["clima"] = np.nan

df_nulos.head()
```

```python
df_nulos.isna()
```

Esto nos permite realizar una visión general pero podemos agregar los valores por columna para entender mejor la información de cada variable:

Suele ser mucho más indicativo poder conocer el % ya que 5 valores nulos en 5 millones de registros es despreciable pero 5 valores nulos en 10 registros es muy significativo. Aprovechando la vectorización de operaciones:

```python
(df_nulos.isna().sum() / df_nulos.shape[0])*100
```

```python
(df_nulos.isna().sum().sum() / df_nulos.size) * 100
```

### Eliminación de valores nulos

Una vez detectados los valores nulos una primera aproximación puede ser eliminarlos. En un posicionamiento radical podríamos plantear eliminar todos los registros que contengan valores nulos. Para ello podemos utilizar el método `.dropna()`:

```python
df_nulos.dropna()
```

En este caso no tiene mucho sentido ya que tenemos una columna en la que todos los valores son nulos. Un primer paso podría ser eliminar esa columna y después podríamos proceder al borrado con este método pero también podemos usar el argumento `thresh` que nos permite definir un umbral para la eliminación de la fila. 

`thresh` indica el número mínimo de valores no nulos para que el registro no se elimine:

```python
df_nulos.dropna(thresh=1)
```

```python
df_nulos.dropna(thresh=8)
```

```python
df_nulos.dropna(thresh=7) #equivalente a solo nulos en clima
```

Análogamente podemos realizar con el mismo método la eliminación por columna utilizando el mismo método `.dropna()` y el argumento `axis=1`:

```python
df_nulos.dropna(axis=1)
```

De nuevo es compatible con el argumento `thresh`:

```python
df_nulos.dropna(axis=1, thresh=3)
```

```python
df_nulos.dropna(axis=1, thresh=23)
```

### Imputación de valores nulos

## 6.5. Sustitución de valores

En esta clase vamos a ver una operación bastante habitual en el tratamiento de datos y especialmente durante su limpieza inicial: la sustitución de valores.

En datasets reales es frecuente encontrar valores que necesitan ser corregidos o estandarizados, por ejemplo:

* Errores tipográficos (Madrid, MAdrid, MADRID, madrid...)
* Formatos diferentes para una misma categoría (hora, h, HORA,hs)
* Códigos que deben convertirse en texto.

El método principal para sustituir valores en Pandas es `.replace()` que recibe como primer argumento el valor antiguo y como segundo el nuevo valor a asignar:

```python
import pandas as pd

data = {
    "id": list(range(1, 31)),
    "nombre": [
        "Ana","Luis","Carmen","Pedro","Lucia","Marta","Jorge","Elena","Raul","Sara",
        "Pablo","Irene","Diego","Laura","Alberto","Nuria","Carlos","Patricia","Sergio","Clara",
        "Adrian","Noelia","Victor","Alicia","Ruben","Diana","Marcos","Paula","Alvaro","Julia"
    ],
    "ciudad": [
        "madrid","Madrid","MADRID","Sevilla","sevilla","SEVILLA","Valencia","valencia","VALENCIA","Barcelona",
        "barcelona","BARCELONA","madrid "," madrid","", "Sevilla","VALENCIA","Madrid","sevilla","Valencia",
        "Barcelona","barcelona","MADRID","valencia","Sevilla","VALENCIA","madrid","Barcelona","sevilla","Madrid"
    ],
    "edad": [
        25,34,29,41,22,38,23,45,36,31,
        27,28,50,33,27,39,44,-5,30,-1,
        42,26,37,48,29,33,31,35,40,28
    ],
    "estado": [
        "activo","Activo","activo","activo","Activo","inactivo","activo","Activo","activo","activo",
        "Activo","inactivo","activo","activo","activo","activo","Activo","activo","activo","activo",
        "activo","Activo","activo","activo","activo","inactivo","Activo","activo","activo","Activo"
    ]
}

df_reemp = pd.DataFrame(data)

df_reemp.head()
```

Para realizar una única modificación podemos indicar simplemente:

```python
df_reemp.replace("madrid", "Madrid")
```

En este caso vemos que necesitamos múltiples sustituciones por lo que también podemos combinar el método con un diccionario de casos:

```python
df_reemp.ciudad.unique()
```

```python
dic_sust = {
    "madrid" : "Madrid",
    "MADRID" : "Madrid",
    "sevilla" : "Sevilla",
    "SEVILLA" : "Sevilla",
    "valencia" : "Valencia",
    "VALENCIA" : "Valencia",
    "BARCELONA" : "Barcelona",
    "barcelona" : "Barcelona",
}

```

```python
df_reemp.replace(dic_sust)
```

Así mismo también podríamos usar un formato de lista para valores concretos, por ejemplo:

```python
df_reemp.replace(["valencia", "VALENCIA"], "Valencia")
```

Otra forma de reemplazo de valores es la sustitución según condiciones. En el caso del ejemplo, vemos que tenemos edades negativas. Esto no tiene ningún sentido por lo que podemos sustituirlas por valores nulos:

```python
df_reemp.loc[df_reemp["edad"] < 0, "edad"] = None
df_reemp
```

Por último en ocasiones resulta útil poder generalizar nuestras sustituciones mediante expresiones regulares. El método `.replace()` nos pemrite utilizar el parámetro `regex=True` para controlar esas sustituciones:

```python
df_reemp.replace(r"(?i)^mad.*", "Madrid", regex=True)
```

La expresión regular se segregaría en:

| Parte  | Significado                  |
| ------ | ---------------------------- |
| `(?i)` | ignora mayúsculas/minúsculas |
| `^`    | inicio                       |
| `mad`  | la palabra empieza por "mad" |

```python
data = {
    "titulo": [
        "Don Quijote", "La Odisea", "Hamlet", "Cien años de soledad",
        "El principito", "La divina comedia", "Romeo y Julieta",
        "El quijote", "La Iliada", "Hamlet ", "Don quijote", "Romancero gitana"
    ],
    
    "autor": [
        "Cervantes", "Homero", "shakespeare", "García Márquez",
        "Saint-Exupéry", "Dante", "shakespeare",
        "cervantes", "HOMERO", "Shakespeare ", "Cervantes", "Lorca"
    ],
    
    "genero": [
        "novela", "epopeya", "teatro", "novela",
        "novela", "poesia", "teatro",
        "novela", "epopeya", "teatro", "novela", "poesia"
    ],
    
    "pais": [
        "España", "Grecia", "Inglaterra", "Colombia",
        "Francia", "Italia", "Inglaterra",
        "españa", "GRECIA", "inglaterra", "España", "España"
    ]
}

df_literatura = pd.DataFrame(data)

df_literatura
```
